`scores.csv` の英語(`english`)の点数を、階級 `[20, 40, 60, 80, 100]` で区切ったとき、**60 点より大きく 80 点以下** の人数を表示してください。

```python
# @starter
import pandas as pd

df = pd.read_csv("scores.csv")
bins = [20, 40, 60, 80, 100]
table = pd.cut(df["english"], bins=bins).value_counts().sort_index()
print(____)
```

```python
# @solution
# @explanation: (60, 80] は3番目の階級なので table.iloc[2] で取り出せます。
import pandas as pd

df = pd.read_csv("scores.csv")
bins = [20, 40, 60, 80, 100]
table = pd.cut(df["english"], bins=bins).value_counts().sort_index()
print(table.iloc[2])
```

```python
# @solution
# @explanation: pandas の var() は既定で不偏分散(ddof=1)を返します。
import pandas as pd

df = pd.read_csv("scores.csv")
var = df["math"].var()
print(round(var, 2))
```

<!-- @question
type: code
unit: 5
-->
`scores.csv` の数学(`math`)と英語(`english`)の点数の **相関係数** を求め、小数第3位まで表示してください。

```python
# @starter
import pandas as pd

df = pd.read_csv("scores.csv")
r = ____
print(round(r, 3))
```

```python
# @solution
# @explanation: Series どうしの相関係数は corr() で求められます。
import pandas as pd

df = pd.read_csv("scores.csv")
r = df["math"].corr(df["english"])
print(round(r, 3))
```
