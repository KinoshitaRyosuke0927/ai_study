# 相関
2つの量的データの関係を、散布図と相関係数で確認します。

## 学習目標
- 散布図から2変数の関係を読み取れる
- 相関係数を計算し、解釈できる
- 相関と因果の違いを説明できる

## 概念

**相関係数** $r$ は、2つの変数の直線的な関係の強さを −1 〜 1 で表します。

- $r$ が 1 に近い:一方が大きいほど、もう一方も大きい(正の相関)
- $r$ が −1 に近い:一方が大きいほど、もう一方は小さい(負の相関)
- $r$ が 0 に近い:直線的な関係はほとんどない

> **注意** 相関があっても、一方が原因でもう一方が起きている(因果関係)とは限りません。

## 例題
勉強時間と数学の点数の関係を散布図で確認します。

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("scores.csv")
plt.scatter(df["study_hours"], df["math"])
plt.xlabel("study_hours")
plt.ylabel("math")
plt.show()
```

```python
df[["math", "english", "study_hours"]].corr().round(2)
```

## 演習

勉強時間(`study_hours`)と英語(`english`)の点数の相関係数を求め、小数第3位まで表示してください。

```python
# @starter
import pandas as pd

df = pd.read_csv("scores.csv")
r = df["study_hours"].____(df["english"])
print(round(r, 3))
```

```python
# @solution
# @title: 勉強時間と英語の相関係数
# @hint: Series どうしの相関係数は corr() で求められます。
# @explanation: 数学より弱いものの、正の相関があります。
import pandas as pd

df = pd.read_csv("scores.csv")
r = df["study_hours"].corr(df["english"])
print(round(r, 3))
```
