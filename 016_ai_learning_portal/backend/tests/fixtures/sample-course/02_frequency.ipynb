# 度数分布とヒストグラム
量的データを階級に分けて数え、分布の形をヒストグラムで確認します。

## 学習目標
- 度数分布表の階級と度数を説明できる
- pandas で度数分布表を作れる
- matplotlib でヒストグラムを描ける

## 概念

量的データは値がばらばらなので、一定の幅の **階級** に区切って、各階級に入るデータの個数(**度数**)を数えます。
これを表にしたものが **度数分布表**、棒グラフにしたものが **ヒストグラム** です。

階級の幅を変えると見え方が変わるため、いくつか試して分布の特徴(山の位置・広がり・偏り)をつかみます。

## 例題
英語の点数を 20 点刻みの階級に分けて数えます。`pd.cut` で各データを階級に割り当てます。

```python
import pandas as pd

df = pd.read_csv("scores.csv")
bins = [20, 40, 60, 80, 100]
pd.cut(df["english"], bins=bins).value_counts().sort_index()
```

同じデータをヒストグラムで表示します。

```python
import matplotlib.pyplot as plt

plt.hist(df["english"], bins=bins, edgecolor="white")
plt.xlabel("english")
plt.ylabel("count")
plt.show()
```
