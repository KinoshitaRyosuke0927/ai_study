学習用のデータで、**性別 `Sex` ごとの年齢の中央値** を求め、その値で年齢の欠損値を埋めてください。性別ごとの中央値と、埋めた後の年齢の平均(小数第2位まで)を表示してください。

```python
# @starter
import pandas as pd

# 学習用のデータを読み込む
train = pd.read_csv("train.csv")

# 性別ごとの年齢の中央値を求めて表示する
age_by_sex = train.groupby("Sex")["Age"].median()
print(age_by_sex)

# 各乗客の性別を中央値に置き換え、年齢の欠損値を埋める(____ に、埋める値の式を書く)
age_filled = train["Age"].fillna(____)

# 埋めた後の年齢の平均を、小数第2位まで表示する
print(round(age_filled.mean(), 2))
```

```python
# @solution
# @explanation: train["Sex"].map(age_by_sex) で、各乗客の性別を性別ごとの中央値に置き換えた列ができ、fillna でそれを使って埋めます。中央値は女性 28 歳、男性 29 歳です。
import pandas as pd

# 学習用のデータを読み込む
train = pd.read_csv("train.csv")

# 性別ごとの年齢の中央値を求めて表示する
age_by_sex = train.groupby("Sex")["Age"].median()
print(age_by_sex)

# 各乗客の性別を中央値に置き換え(map)、年齢の欠損値を埋める
age_filled = train["Age"].fillna(train["Sex"].map(age_by_sex))

# 埋めた後の年齢の平均を、小数第2位まで表示する
print(round(age_filled.mean(), 2))
```

学習用のデータで、乗船した港 `Embarked`(欠損値は "S" で埋める)をワンホットエンコーディングし(列名の先頭は `Embarked`)、**各列の 1 の数** を表示してください。

```python
# @starter
import pandas as pd

# 学習用のデータを読み込む
train = pd.read_csv("train.csv")

# 欠損値を "S" で埋めてから、港ごとに 0 / 1 の列を作る(____ に、埋める処理の式を書く)
dummies = pd.get_dummies(____, prefix="Embarked", dtype=int)

# 各列の 1 の数を表示する
print(dummies.sum())
```

```python
# @solution
# @explanation: 欠損値を S で埋めたので、Embarked_S の 1 の数は 614 + 2 = 616 になります。各行でどれか1つの列だけが 1 なので、3 列の合計は 891(全員の人数)になります。
import pandas as pd

# 学習用のデータを読み込む
train = pd.read_csv("train.csv")

# 欠損値を "S" で埋めてから、港ごとに 0 / 1 の列を作る
dummies = pd.get_dummies(train["Embarked"].fillna("S"), prefix="Embarked", dtype=int)

# 各列の 1 の数を表示する
print(dummies.sum())
```

<!-- @question

answer: 予測用のデータにない値の列が作られず、学習用と列が合わなくなることがあるため。
unit: 3
explanation: get_dummies はデータに含まれている値の列しか作りません。予測用のデータに学習用にある値が含まれていないと、その列が作られず、列の数や順番が学習用と合わなくなります。reindex(columns=学習用の列名, fill_value=0) でそろえます。
-->
学習用・予測用それぞれに `pd.get_dummies` を使った後、`reindex(columns=学習用の列名, fill_value=0)` で予測用の列をそろえる理由として最も適切なものを選べ。

- get_dummies は欠損値を作るので、0 で埋める必要があるため。
- 予測用のデータにない値の列が作られず、学習用と列が合わなくなることがあるため。
- 予測用のデータの行の順番を、学習用と同じにするため。
- reindex をしないと、get_dummies が文字列の列を返すため。

学習用のデータで、家族の人数 `FamilySize`(`SibSp + Parch + 1`)が **5 人以上** かどうかを表す `LargeFamily` 列(5 人以上なら 1、それ以外は 0)を作り、`LargeFamily` ごとの生存率を小数第3位まで表示してください。

```python
# @starter
import pandas as pd

# 学習用のデータを読み込み、家族の人数の列を作る
train = pd.read_csv("train.csv")
train["FamilySize"] = train["SibSp"] + train["Parch"] + 1

# 5 人以上なら 1、それ以外は 0 の列を作る(____ に、5 人以上かどうかを判定する比較の式を書く)
train["LargeFamily"] = (____).astype(int)

# LargeFamily ごとの生存率を、小数第3位まで表示する
print(train.groupby("LargeFamily")["Survived"].mean().round(3))
```

```python
# @solution
# @explanation: >= は「以上」の比較です。5 人以上の家族の生存率は約 18% で、それ以外(約 39%)より大きく下がります。
import pandas as pd

# 学習用のデータを読み込み、家族の人数の列を作る
train = pd.read_csv("train.csv")
train["FamilySize"] = train["SibSp"] + train["Parch"] + 1

# 家族の人数が 5 以上なら True → astype(int) で 1 / 0 に変える
train["LargeFamily"] = (train["FamilySize"] >= 5).astype(int)

# LargeFamily ごとの生存率を、小数第3位まで表示する
print(train.groupby("LargeFamily")["Survived"].mean().round(3))
```
