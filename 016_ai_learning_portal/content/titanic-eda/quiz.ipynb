学習用のデータで、年齢 `Age` を **人数がほぼ等しい 4 つの区間** に分け、区間ごとの生存率を小数第3位まで表示してください。

```python
# @starter
import pandas as pd

# 学習用のデータを読み込む
train = pd.read_csv("train.csv")

# 年齢を、人数がほぼ等しい 4 つの区間に分ける(____ に、区間に分ける式を書く)
age_group = ____

# 区間ごとの生存率を、小数第3位まで表示する
print(train.groupby(age_group, observed=True)["Survived"].mean().round(3))
```

```python
# @solution
# @explanation: pd.qcut(train["Age"], 4) で四分位に分けます。どの区間も生存率は約 36〜43% で、四分位の区切りでは差が小さく見えます。子どもの高い生存率は、最も若い区間(21 歳以下)の中に薄まってしまうためです。区切り方によって見え方が変わることに注意しましょう。
import pandas as pd

# 学習用のデータを読み込む
train = pd.read_csv("train.csv")

# 年齢を、人数がほぼ等しい 4 つの区間(四分位)に分ける
age_group = pd.qcut(train["Age"], 4)

# 区間ごとの生存率を、小数第3位まで表示する
print(train.groupby(age_group, observed=True)["Survived"].mean().round(3))
```
