# 決定係数と F 検定
重回帰モデル全体の当てはまりを、決定係数・自由度調整済み決定係数と、分散分析表の F 検定で評価します。説明変数を増やしたときの決定係数の振る舞いと、モデルの比べ方を身につけます。

## 学習目標
- 重回帰の分散分析表を作り、決定係数と F 値を求められる
- 自由度調整済み決定係数を計算し、決定係数との違いを説明できる
- F 検定の仮説を述べ、判断できる
- 説明変数の数が異なるモデルを、適切な指標で比べられる

## 重回帰の分散分析表

単回帰と同じく、全変動は回帰変動と残差変動に分けられます($S_T = S_R + S_E$)。説明変数が $p$ 個のとき、分散分析表は次のようになります。

| 要因 | 平方和 | 自由度 | 平均平方 | F 値 |
|---|---|---|---|---|
| 回帰 | $S_R$ | $p$ | $V_R = S_R / p$ | $F = V_R / V_E$ |
| 残差 | $S_E$ | $n - p - 1$ | $V_E = S_E / (n - p - 1)$ | |
| 全体 | $S_T$ | $n - 1$ | | |

**決定係数** は単回帰と同じく $R^2 = S_R / S_T = 1 - S_E / S_T$ です(重回帰では **重相関係数** $R$ の2乗。$R$ は $y$ と予測値 $\hat{y}$ の相関係数)。

## F 検定(モデル全体の検定)

分散分析表の F 値は、次の検定に使います。

- 帰無仮説 $H_0: \beta_1 = \beta_2 = \dots = \beta_p = 0$(どの説明変数も役に立たない)
- 対立仮説 $H_1$:少なくとも1つの $\beta_j$ は 0 でない
- $H_0$ のもとで $F$ は **自由度 $(p,\ n - p - 1)$ の F 分布** に従う。$F > F_{0.05}(p, n - p - 1)$ なら 5% で棄却

**例**:$n = 30$、$p = 2$、$S_T = 500$、$S_E = 150$ のとき

| 要因 | 平方和 | 自由度 | 平均平方 | F 値 |
|---|---|---|---|---|
| 回帰 | 350 | 2 | 175 | 31.5 |
| 残差 | 150 | 27 | 5.556 | |
| 全体 | 500 | 29 | | |

$R^2 = 350 \div 500 = 0.70$。$F_{0.05}(2, 27) = 3.354$ より $F = 31.5$ は大きいので、$H_0$ を棄却します。

## 自由度調整済み決定係数

決定係数 $R^2$ には、**説明変数を加えると(役に立たない変数でも)必ず増える** という性質があります。そのため、説明変数の数が異なるモデルを $R^2$ で比べると、変数が多いモデルが常に有利になります。

そこで、平方和を自由度で割ってから比べるのが **自由度調整済み決定係数** です。

$$R^{*2} = 1 - \frac{S_E / (n - p - 1)}{S_T / (n - 1)} = 1 - (1 - R^2)\frac{n - 1}{n - p - 1}$$

**例**(上の分散分析表):$R^{*2} = 1 - (150 / 27) \div (500 / 29) = 1 - 5.556 \div 17.24 \approx 0.678$

次の図は、中古マンションの例(広さ・築年数)に、価格と無関係な乱数の変数を1つずつ加えていったときの変化です。$R^2$ は増え続けますが、$R^{*2}$ は下がっていきます。

![無関係な変数を加えたときの決定係数と自由度調整済み決定係数](images/u04_adj_r2.png)

- $R^{*2} \le R^2$。$R^{*2}$ は負になることもある
- 説明変数の数が異なるモデルを比べるときは、$R^2$ ではなく $R^{*2}$(や AIC などの指標)を使う

## まとめ

- 重回帰の分散分析表:回帰の自由度 $p$、残差の自由度 $n - p - 1$。$F = V_R / V_E$。
- F 検定の $H_0$:すべての偏回帰係数が 0。棄却されれば「少なくとも1つは 0 でない」。
- 決定係数 $R^2 = 1 - S_E/S_T$ は、説明変数を加えると必ず増える。
- 自由度調整済み決定係数 $R^{*2} = 1 - \{S_E/(n-p-1)\}/\{S_T/(n-1)\}$ で、説明変数の数が異なるモデルを比べる。
