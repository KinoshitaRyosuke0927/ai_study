# 一元配置分散分析
3 つ以上の群の平均に差があるかを調べる一元配置分散分析を学びます。全体のばらつきを群間と群内に分け、分散分析表を手計算で作れるようにします。

## 学習目標
- 実験計画の用語(因子・水準・繰り返し)を説明できる
- 一元配置のモデルと、分散分析の仮説を述べられる
- 群間平方和・群内平方和と自由度を求め、分散分析表を作れる
- F 値から判断し、結論を正しく述べられる

## 実験計画と分散分析

「データ収集と確率・確率分布」の講座では、フィッシャーの3原則(反復・無作為化・局所管理)に沿って実験を計画することを学びました。この単元と次の単元では、計画した実験で得たデータを **分散分析** で解析します。

| 用語 | 意味 | 例(肥料の比較) |
|---|---|---|
| **因子** | 結果に影響すると考えて取り上げる要因 | 肥料の種類 |
| **水準** | 因子の値・種類 | 肥料 A・B・C の 3 水準 |
| **繰り返し** | 同じ水準で行う実験の回数 | 各肥料で 4 区画 |

取り上げる因子が1つの実験を **一元配置**、2つの実験を **二元配置** といいます。

なぜ「2群の t 検定」をくり返さないのか:3 群なら A と B、A と C、B と C の 3 回の検定を行うことになり、どれかで誤って有意となる確率(第1種の過誤)が 5% より大きくなってしまいます。分散分析は、すべての群をまとめて1回の検定で判断します。

## 一元配置のモデルと仮説

因子の水準が $k$ 個、水準 $i$ の $j$ 番目の観測値を $y_{ij}$ とすると

$$y_{ij} = \mu + \alpha_i + \varepsilon_{ij}, \qquad \varepsilon_{ij} \sim N(0, \sigma^2)\ \text{(独立)}$$

$\mu$ は全体の平均、$\alpha_i$ は水準 $i$ の効果(全体の平均からのずれ)です。誤差は回帰と同じく、平均 0・**等分散**・独立・正規分布を仮定します。

- 帰無仮説 $H_0: \alpha_1 = \alpha_2 = \dots = \alpha_k = 0$(すべての水準の母平均が等しい)
- 対立仮説 $H_1$:少なくとも1つの水準の母平均が異なる

## 変動の分解と分散分析表

全体の平均を $\bar{y}$、水準 $i$ の平均を $\bar{y}_i$、水準 $i$ のデータ数を $n_i$、全データ数を $N$ とします。

$$\underbrace{\sum_{i}\sum_{j} (y_{ij} - \bar{y})^2}_{\text{総平方和 } S_T} = \underbrace{\sum_{i} n_i(\bar{y}_i - \bar{y})^2}_{\text{群間平方和 } S_A} + \underbrace{\sum_{i}\sum_{j} (y_{ij} - \bar{y}_i)^2}_{\text{群内平方和 } S_E}$$

| 要因 | 平方和 | 自由度 | 平均平方 | F 値 |
|---|---|---|---|---|
| 群間(因子 A) | $S_A$ | $k - 1$ | $V_A = S_A / (k - 1)$ | $F = V_A / V_E$ |
| 群内(誤差) | $S_E$ | $N - k$ | $V_E = S_E / (N - k)$ | |
| 全体 | $S_T$ | $N - 1$ | | |

$H_0$ のもとで $F$ は自由度 $(k - 1,\ N - k)$ の F 分布に従います。群内のばらつきに比べて群間のばらつきが大きいほど、$F$ は大きくなります。

**例**:3 種類の肥料(各 4 区画)の収穫量

| 肥料 | データ | 平均 |
|---|---|---|
| A | 5, 7, 6, 6 | 6 |
| B | 9, 11, 10, 10 | 10 |
| C | 7, 8, 9, 8 | 8 |

全体の平均は 8 です。

![群間のばらつき(群平均 − 全体平均)と群内のばらつき(値 − 群平均)](images/u06_oneway.png)

- 群間平方和 $S_A = 4 \times \{(6-8)^2 + (10-8)^2 + (8-8)^2\} = 4 \times 8 = 32$
- 群内平方和 $S_E$:A は $1 + 1 + 0 + 0 = 2$、B も 2、C も 2 で、合計 6

| 要因 | 平方和 | 自由度 | 平均平方 | F 値 |
|---|---|---|---|---|
| 肥料 | 32 | 2 | 16 | 24.0 |
| 誤差 | 6 | 9 | 0.667 | |
| 全体 | 38 | 11 | | |

$F_{0.05}(2, 9) = 4.256$ より $F = 24.0$ は大きいので、$H_0$ を棄却し、肥料によって平均収穫量に差があると判断します。

**注意**:分散分析で有意になっても、わかるのは「どこかに差がある」ことまでです。どの水準の間に差があるかは、**多重比較**(ボンフェローニ法、テューキー法など)で調べます。

4 種類の学習方法(各 5 人、合計 20 人)でテストの点数を比べたところ、各方法の平均点は 12, 15, 14, 19 点、群内平方和は 96 であった。次の問いに答えよ。

有意水準 5% での判断として最も適切なものを選べ。ただし、$F_{0.05}(3, 16) = 3.239$ である。

- F₀.₀₅(3, 16) より大きいので、学習方法によって平均点に差があると判断する。
- F₀.₀₅(3, 16) より大きいので、4 つの学習方法の平均点はすべて互いに異なると判断する。
- F₀.₀₅(3, 16) より大きいので、学習方法の間に差はないと判断する。
- 平均点が最も高い方法と最も低い方法の差が 7 点なので、F 値を使わなくても差があると判断できる。

4 つの群の平均を比べるとき、2 群の t 検定をくり返すのではなく一元配置分散分析を使う理由として最も適切なものを選べ。

- 2 群の t 検定をくり返すと、全体として第1種の過誤を起こす確率が有意水準より大きくなるから。
- 2 群の t 検定は、群の大きさが等しいときには使えないから。
- 分散分析は、誤差が正規分布に従わなくても使えるから。
- 分散分析は、どの群の間に差があるかを1回で特定できるから。

## まとめ

- 因子・水準・繰り返し。因子が1つの実験が一元配置。
- 総平方和 = 群間平方和 + 群内平方和。自由度は $k - 1$ と $N - k$。
- $F = V_A / V_E$ が自由度 $(k-1, N-k)$ の F 分布の上側 5% 点より大きければ、「少なくとも1つの水準の母平均が異なる」。
- どこに差があるかは多重比較で調べる。t 検定のくり返しは第1種の過誤が大きくなる。
