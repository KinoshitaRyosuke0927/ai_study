# 多重共線性とダミー変数
重回帰で説明変数どうしの相関が強いときに起こる「多重共線性」と、その目安となる VIF を学びます。また、立地や性別などの質的な変数を回帰に入れる「ダミー変数」の作り方と、係数の読み方を身につけます。

## 学習目標
- 多重共線性で起こる問題を説明できる
- VIF を計算し、多重共線性の有無を判断できる
- 質的変数をダミー変数で表し、必要なダミー変数の数を答えられる
- ダミー変数の係数を、基準のカテゴリとの差として解釈できる

## 多重共線性

説明変数どうしの相関が非常に強いことを **多重共線性**(マルチコ)といいます。多重共線性があると、次のような問題が起こります。

| 症状 | 理由 |
|---|---|
| 偏回帰係数の **標準誤差が大きく** なり、t 値が小さく(有意になりにくく)なる | 相関の強い変数どうしでは、「他の変数を一定にしたとき」の変化を見分けにくい |
| データを少し変えるだけで、係数が大きく変わる | 推定が不安定 |
| 係数の符号が、常識や単回帰の結果と逆になることがある | |
| F 検定は有意(全体としては予測に役立つ)なのに、個々の係数は有意でない | |

多重共線性があっても、**予測値そのものは** 大きくは影響を受けません。問題になるのは、個々の係数の解釈です。

## VIF(分散拡大要因)

説明変数 $x_j$ を、**他の説明変数で回帰したときの決定係数** を $R_j^2$ とします。

$$\mathrm{VIF}_j = \frac{1}{1 - R_j^2}$$

- $x_j$ が他の変数とまったく相関しなければ $R_j^2 = 0$ で、VIF = 1
- VIF は、多重共線性によって $b_j$ の分散が何倍に拡大しているかを表す
- **VIF > 10**($R_j^2 > 0.9$)を、多重共線性を疑う目安とすることが多い

**例**:中古マンションの例では、広さと築年数の相関係数が 0.92 でした。

![広さと築年数の散布図](images/u05_collinearity.png)

広さを築年数で回帰したときの決定係数は $0.922^2 \approx 0.851$ なので、$\mathrm{VIF} = 1 / (1 - 0.851) \approx 6.7$。目安の 10 には届きませんが、係数の標準誤差は相関がない場合の約 6.7 倍の分散(標準誤差で約 2.6 倍)になっています。

**対処法**:相関の強い変数の一方を除く、変数をまとめた合成変数を使う、データを増やす、など。

## ダミー変数

立地(駅前・商業施設内・郊外)のような **質的変数** は、0 か 1 の値をとる **ダミー変数** で回帰に入れます。

| 立地 | $D_1$(駅前) | $D_2$(商業施設内) |
|---|---|---|
| 駅前 | 1 | 0 |
| 商業施設内 | 0 | 1 |
| 郊外(基準) | 0 | 0 |

- **カテゴリが $k$ 個なら、ダミー変数は $k - 1$ 個**。すべてが 0 のカテゴリが **基準** になる
- $k$ 個すべてを入れると、ダミー変数の和が常に 1(切片と同じ)になり、係数が求められない(完全な多重共線性)
- ダミー変数の係数は、**他の変数が同じときの、基準のカテゴリとの差**

**例**:店舗の売上 $y$ を、面積 $x$ と立地で回帰して $\hat{y} = 150 + 2.0x + 40D_1 + 15D_2$ が得られたとします。

![ダミー変数を入れた回帰(立地ごとに平行な直線)](images/u05_dummy.png)

- 面積が同じなら、駅前の店舗は郊外より平均 40、商業施設内は郊外より平均 15 売上が多い
- 駅前と商業施設内の差は $40 - 15 = 25$
- 面積 30 の駅前の店舗の予測値は $150 + 2.0 \times 30 + 40 = 250$、郊外なら $210$

重回帰分析の結果のうち、多重共線性が疑われる状況として最も適切なものを選べ。

- すべての説明変数の VIF が 1 に近い。
- 分散分析表の F 検定は有意だが、個々の偏回帰係数の t 検定はどれも有意でない。
- 残差プロットで、予測値が大きいほど残差のばらつきが大きい。
- 決定係数が 0.3 と小さい。

## まとめ

- 多重共線性:説明変数どうしの相関が強い。係数の標準誤差が大きくなり、係数が不安定になる(予測値への影響は小さい)。
- $\mathrm{VIF}_j = 1/(1 - R_j^2)$。10 を超えると多重共線性を疑う。
- 質的変数は $k - 1$ 個のダミー変数で表す。すべて 0 のカテゴリが基準。
- ダミー変数の係数は、他の変数が同じときの基準カテゴリとの差。
