# 重回帰と偏回帰係数
複数の説明変数を使う重回帰モデルと、その係数(偏回帰係数)の意味を学びます。単回帰と重回帰で係数が大きく変わる理由を理解し、統計ソフトの出力を読み取れるようにします。

## 学習目標
- 重回帰モデルの式と、残差の自由度を説明できる
- 偏回帰係数を「他の変数を一定にしたとき」の関係として解釈できる
- 単回帰の係数と偏回帰係数が異なる理由を説明できる
- 重回帰の出力から予測値を求め、各係数の t 検定を判断できる

## 重回帰モデル

$p$ 個の説明変数 $x_1, x_2, \dots, x_p$ で $y$ を説明するモデルを **重回帰モデル** といいます。

$$y_i = \beta_0 + \beta_1 x_{1i} + \beta_2 x_{2i} + \dots + \beta_p x_{pi} + \varepsilon_i$$

- 誤差 $\varepsilon_i$ の仮定は単回帰と同じ(平均 0・等分散・独立・正規分布)
- 係数は、単回帰と同じく **残差平方和を最小にする** ように求める(最小二乗法)
- 推定する係数は切片を含めて $p + 1$ 個なので、**残差の自由度は $n - p - 1$**。誤差分散の推定値は $s^2 = S_E/(n - p - 1)$
- 各係数の t 検定($H_0: \beta_j = 0$)は、自由度 $n - p - 1$ の t 分布を使う

## 偏回帰係数の意味

重回帰の係数 $\beta_j$ を **偏回帰係数** といいます。

> 偏回帰係数 $\beta_j$ は、**他の説明変数の値を一定にしたまま** $x_j$ を 1 増やしたときの、$y$ の平均的な変化量

単回帰の係数は、他の変数の違いの影響も含んだ「見かけの関係」です。説明変数どうしに相関があると、両者は大きく異なることがあります。

**例**:中古マンション 40 件の価格(万円)を、広さ(m²)と築年数(年)で説明します。この 40 件では、広い物件ほど築年数が古い傾向があります(広さと築年数の相関係数 0.92)。

| モデル | 築年数の係数 | 解釈 |
|---|---|---|
| 単回帰(価格 ~ 築年数) | **+23.6** | 古い物件ほど高い?(古い物件は広いので、広さの効果が混ざっている) |
| 重回帰(価格 ~ 広さ + 築年数) | **−48.5** | 広さが同じなら、築年数が 1 年古いと価格は平均約 48.5 万円安い |

![単回帰と重回帰で築年数の係数の符号が変わる](images/u03_partial.png)

右の図は、広さで 3 つのグループに分けたものです。広さがほぼ同じグループの中で見ると、古い物件ほど安くなっています。

## 重回帰の出力を読む

この例の統計ソフトの出力の一部です($n = 40$、$p = 2$)。

```
               coef    std err        t    P>|t|
const      763.8489    226.173    3.377    0.002
area        47.5996      5.446    8.741    0.000
age        -48.4737      8.934   -5.426    0.000

R-squared:  0.769    Adj. R-squared:  0.756
F-statistic: 61.55   No. Observations:  40
```

- 回帰式は $\hat{y} = 763.8 + 47.60 \times \text{広さ} - 48.47 \times \text{築年数}$
- 広さ 70 m²・築 20 年の物件の予測価格は $763.8 + 47.60 \times 70 - 48.47 \times 20 \approx 3126$ 万円
- 残差の自由度は $40 - 2 - 1 = 37$。t 値の判断には $t_{0.025}(37) \approx 2.03$ を使う。広さ・築年数とも $|t|$ がこれより大きく、5% で有意

前の問題の重回帰分析で、説明変数 $x_2$ の偏回帰係数は 6.3、その標準誤差は 1.8 であった。有意水準 5% の両側検定($H_0: \beta_2 = 0$)の判断として正しいものを選べ。ただし、$t_{0.025}(45) = 2.014$ である。

- t = 3.5 で、t₀.₀₂₅(45) = 2.014 より大きいので有意である。
- t = 3.5 で、t₀.₀₂₅(45) = 2.014 より大きいので有意でない。
- t = 0.29 で、t₀.₀₂₅(45) = 2.014 より小さいので有意でない。
- 重回帰では、個々の偏回帰係数を検定することはできない。

answer: 気温が同じ日どうしで比べると、降水量が 1 mm 多い日は来店者数が平均的に 3.2 人少ない。

ある店舗の1日の来店者数 $y$ を、最高気温 $x_1$(℃)と降水量 $x_2$(mm)で重回帰分析したところ、$\hat{y} = 180 + 4.5x_1 - 3.2x_2$ となった。降水量の係数の解釈として最も適切なものを選べ。

- 気温を考えないとき、降水量が 1 mm 多い日は来店者数が平均的に 3.2 人少ない。
- 気温が同じ日どうしで比べると、降水量が 1 mm 多い日は来店者数が平均的に 3.2 人少ない。
- 降水量を 1 mm 減らせば、来店者数を 3.2 人増やせる。
- 降水量の係数の絶対値は気温の係数より小さいので、降水量の影響は気温より小さい。

```
            coef   std err       t    P>|t|
const    52.0000     8.000   6.500    0.000
x1        2.4000     0.600   4.000    0.000
x2       -0.9000     0.750  -1.200    0.241
```

この出力について述べた次の記述 I〜III のうち、正しいものの組み合わせを選べ。

I. 有意水準 5% で、$x_1$ の偏回帰係数は 0 でないと判断できる。

II. $x_2$ の係数は負なので、有意水準 5% で、$x_2$ が大きいほど $y$ は小さいと判断できる。

III. 各係数の t 検定には、自由度 27 の t 分布を使う。

- I のみ
- II のみ
- I と II のみ
- I と III のみ
- I と II と III

## まとめ

- 重回帰モデル $y = \beta_0 + \beta_1 x_1 + \dots + \beta_p x_p + \varepsilon$。残差の自由度は $n - p - 1$。
- 偏回帰係数は「他の説明変数を一定にしたとき」の $y$ の平均的な変化量。
- 説明変数どうしに相関があると、単回帰の係数と偏回帰係数は大きく異なり、符号が変わることもある。
- 係数の大きさは単位に依存する。回帰係数は因果関係を保証しない。
