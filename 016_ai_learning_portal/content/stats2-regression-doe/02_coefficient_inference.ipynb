# 回帰係数の検定と区間推定
回帰係数の推定値とその標準誤差から、「$x$ は $y$ の予測に役立つか」を t 検定で判断し、回帰係数の信頼区間を求めます。統計ソフトの係数表の読み方も身につけます。

## 学習目標
- 回帰係数の t 検定の仮説・検定統計量・自由度を説明し、判断できる
- 回帰係数の信頼区間を求められる
- 単回帰で t 値と分散分析表の F 値の関係($F = t^2$)を使える
- 統計ソフトの係数表(coef・std err・t・p 値・信頼区間)を読み取れる

## 回帰係数の t 検定

「$x$ が $y$ に(直線的な)影響を持たない」ことは、$\beta_1 = 0$ と表せます。そこで次の検定を行います。

- 帰無仮説 $H_0: \beta_1 = 0$、対立仮説 $H_1: \beta_1 \neq 0$
- 検定統計量 $t = \dfrac{b}{\mathrm{se}(b)}$。$H_0$ のもとで **自由度 $n - 2$ の t 分布** に従う
- $|t| > t_{0.025}(n - 2)$ なら、有意水準 5% で $H_0$ を棄却する

**例**:5 組のデータ($b = 0.6$、$\mathrm{se}(b) \approx 0.283$)では

$$t = 0.6 \div 0.283 \approx 2.12$$

自由度 $5 - 2 = 3$ の t 分布の上側 2.5% 点は $t_{0.025}(3) = 3.182$ です。$|t| = 2.12 < 3.182$ なので、有意水準 5% で $H_0$ は棄却できません(p 値は約 0.124)。データが 5 組と少ないため、傾きが 0 でないとはいえません。

![回帰係数の t 検定](images/u02_t_test.png)

切片についても、$t = a / \mathrm{se}(a)$ で同じように $\beta_0 = 0$ を検定できます(ただし、実務で関心があるのは多くの場合、回帰係数の方です)。

## 回帰係数の信頼区間

$\beta_1$ の 95% 信頼区間は

$$b \pm t_{0.025}(n - 2) \times \mathrm{se}(b)$$

**例**:5 組のデータでは $0.6 \pm 3.182 \times 0.283 = 0.6 \pm 0.900$ なので、95% 信頼区間は $[-0.30,\ 1.50]$。区間が **0 を含む** ことは、両側 5% の検定で $H_0: \beta_1 = 0$ が棄却されないことと対応しています。

## t 値と F 値の関係

単回帰では、回帰係数の t 検定と、分散分析表の F 検定は同じ検定です。

$$F = t^2$$

**例**:5 組のデータでは $t^2 = 2.121^2 = 4.5$。「記述統計」の講座で作った分散分析表の F 値 $3.6 \div 0.8 = 4.5$ と一致します。

## 統計ソフトの係数表を読む

統計ソフトの回帰分析の出力では、係数ごとに次の値が並びます。

| 列 | 意味 |
|---|---|
| `coef` | 推定値($a$ や $b$) |
| `std err` | 標準誤差 |
| `t` | t 値 = coef ÷ std err |
| `P>\|t\|` | 両側検定の p 値($\beta = 0$ の検定) |
| `[0.025  0.975]` | 95% 信頼区間の下限・上限 |

「記述統計」の講座の出力(生徒 120 人、数学の点数を学習時間で回帰)では、学習時間の t 値は $2.1528 \div 0.241 \approx 8.93$ でした。自由度 $120 - 2 = 118$ の t 分布の上側 2.5% 点(約 1.98)より大きいので、学習時間の係数は有意水準 5% で有意です。

```
            coef   std err       t    P>|t|    [0.025    0.975]
const    12.4000     3.100   4.000    0.001     5.887    18.913
x         0.8500     0.250   3.400    0.003     0.325     1.375
```

この出力について述べた次の記述 I〜III のうち、正しいものの組み合わせを選べ。

I. 有意水準 5% の両側検定で、x の回帰係数は 0 でないと判断できる。

II. x の回帰係数の 95% 信頼区間は 0 を含む。

III. x の t 値は、coef を std err で割った値である。

- I のみ
- II のみ
- I と II のみ
- I と III のみ
- I と II と III

## まとめ

- $H_0: \beta_1 = 0$ の検定:$t = b/\mathrm{se}(b)$、自由度 $n - 2$ の t 分布。
- 95% 信頼区間 $b \pm t_{0.025}(n-2)\,\mathrm{se}(b)$。0 を含まないことと、5% で有意であることが対応する。
- 単回帰では $F = t^2$(分散分析表の F 検定と回帰係数の t 検定は同じ)。
- 係数表の coef・std err・t・p 値・信頼区間の関係を押さえる。
