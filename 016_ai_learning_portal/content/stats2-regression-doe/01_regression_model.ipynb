# 単回帰モデルと回帰係数の推定
「記述統計」の講座では、データに当てはまる回帰直線を最小二乗法で求めました。この講座では、データの背後に「母集団の回帰直線」があると考える **回帰モデル** を導入し、回帰係数の推定値がどれくらいばらつくか(標準誤差)を求めます。

## 学習目標
- 単回帰モデルの式と、誤差に置く仮定を説明できる
- 誤差と残差の違いを説明できる
- 残差から誤差分散の推定値を求められる
- 回帰係数・切片の標準誤差を計算できる

## 単回帰モデル

$x$ の値を決めたとき、$y$ は「直線の上の値」に誤差が加わって観測されると考えます。

$$y_i = \beta_0 + \beta_1 x_i + \varepsilon_i \qquad (i = 1, 2, \dots, n)$$

- $\beta_0$, $\beta_1$:**母回帰係数**(母集団の切片と傾き)。未知の定数
- $\varepsilon_i$:**誤差**。誤差について次の仮定を置きます

| 仮定 | 内容 |
|---|---|
| 平均 0 | $E[\varepsilon_i] = 0$(誤差は平均的に直線の上下に均等に散らばる) |
| 等分散 | $V[\varepsilon_i] = \sigma^2$($x$ によらず同じばらつき) |
| 独立 | $\varepsilon_1, \dots, \varepsilon_n$ は互いに独立 |
| 正規性 | $\varepsilon_i \sim N(0, \sigma^2)$(検定・区間推定を行うときに必要) |

この仮定のもとでは、$x$ を決めたときの $y$ は、平均 $\beta_0 + \beta_1 x$、分散 $\sigma^2$ の正規分布に従います。$x$ は確率変数ではなく、与えられた値として扱います。

![単回帰モデル:各 x での y の分布](images/u01_model.png)

**例**:$y = 10 + 2x + \varepsilon$、$\varepsilon \sim N(0, 3^2)$ のモデルでは、$x = 4$ のときの $y$ は $N(18, 3^2)$ に従います。

## 推定値・誤差・残差

母回帰係数 $\beta_0$, $\beta_1$ は、データから最小二乗法で推定します(「記述統計」の講座で学んだ $a$, $b$)。

$$\hat{\beta}_1 = b = \frac{S_{xy}}{S_{xx}}, \qquad \hat{\beta}_0 = a = \bar{y} - b\bar{x}$$

ここで $S_{xx} = \sum (x_i - \bar{x})^2$($x$ の偏差平方和)、$S_{xy} = \sum (x_i - \bar{x})(y_i - \bar{y})$(偏差の積の和)です。

| 量 | 意味 | 観測できるか |
|---|---|---|
| 誤差 $\varepsilon_i = y_i - (\beta_0 + \beta_1 x_i)$ | 母回帰直線からのずれ | できない($\beta_0$, $\beta_1$ が未知) |
| 残差 $e_i = y_i - (a + b x_i)$ | 推定した回帰直線からのずれ | できる |

最小二乗推定量 $a$, $b$ は、それぞれ $\beta_0$, $\beta_1$ の **不偏推定量** です。

## 誤差分散の推定と標準誤差

誤差の分散 $\sigma^2$ は、残差平方和 $S_E = \sum e_i^2$ を **自由度 $n - 2$** で割って推定します($a$ と $b$ の 2 つを推定しているので 2 を引く)。

$$s^2 = \frac{S_E}{n - 2} \qquad (\text{分散分析表の残差の平均平方 } V_E \text{ と同じ})$$

回帰係数 $b$ と切片 $a$ の **標準誤差**(推定値のばらつきの大きさ)は、$\sigma$ を $s$ で置き換えて次のように求めます。

$$\mathrm{se}(b) = \frac{s}{\sqrt{S_{xx}}}, \qquad \mathrm{se}(a) = s\sqrt{\frac{1}{n} + \frac{\bar{x}^2}{S_{xx}}}$$

- $S_{xx}$ が大きい(**$x$ を広い範囲でとる**、データが多い)ほど、$b$ の標準誤差は小さくなる
- 誤差のばらつき $s$ が小さいほど、標準誤差は小さくなる

**例**:「記述統計」の講座の 5 組のデータ($x$ = 1〜5、$y$ = 2, 4, 5, 4, 5)

- $b = 0.6$、$a = 2.2$、$S_{xx} = 10$、$\bar{x} = 3$、残差平方和 $S_E = 2.4$
- $s^2 = 2.4 \div (5 - 2) = 0.8$
- $\mathrm{se}(b) = \sqrt{0.8} \div \sqrt{10} = \sqrt{0.08} \approx 0.283$
- $\mathrm{se}(a) = \sqrt{0.8 \times (1/5 + 9/10)} = \sqrt{0.88} \approx 0.938$

## まとめ

- 単回帰モデル $y_i = \beta_0 + \beta_1 x_i + \varepsilon_i$。誤差は平均 0・等分散・独立(検定では正規分布)を仮定する。
- 誤差は観測できない。残差は推定した回帰直線からのずれで、データから計算できる。
- 誤差分散の推定値 $s^2 = S_E/(n-2)$。
- $\mathrm{se}(b) = s/\sqrt{S_{xx}}$。$x$ を広い範囲でとるほど、$b$ を精度よく推定できる。
