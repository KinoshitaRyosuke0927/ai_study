# ベイズの定理
結果(検査で陽性だった、不良品だった)から、その原因の確率を求めるベイズの定理を学びます。全確率の定理と組み合わせて、手計算で使えるようにします。

## 学習目標
- 全確率の定理を使って、いくつかの原因を通じて起こる事象の確率を求められる
- ベイズの定理を使って、結果から原因の確率(事後確率)を求められる
- 事前確率・事後確率の意味を説明できる
- 「陽性的中率が意外に低い」理由を、人数に置き換えて説明できる

## 全確率の定理

事象 $B_1, B_2, \dots, B_k$ が互いに排反で、合わせると全体になる(どれか1つが必ず起こる)とき、事象 $A$ の確率は

$$P(A) = P(B_1)P(A \mid B_1) + P(B_2)P(A \mid B_2) + \cdots + P(B_k)P(A \mid B_k)$$

と、原因ごとに分けて足し合わせて求められます。

## ベイズの定理

$A$ が起こったとき、その原因が $B_i$ である確率は

$$P(B_i \mid A) = \frac{P(B_i)\,P(A \mid B_i)}{P(A)} = \frac{P(B_i)\,P(A \mid B_i)}{\sum_j P(B_j)\,P(A \mid B_j)}$$

| 用語 | 意味 |
|---|---|
| **事前確率** $P(B_i)$ | 結果を知る前の、原因 $B_i$ の確率 |
| **事後確率** $P(B_i \mid A)$ | 結果 $A$ を知った後の、原因 $B_i$ の確率 |

ベイズの定理は、「原因 → 結果」の確率 $P(A \mid B_i)$ から、「結果 → 原因」の確率 $P(B_i \mid A)$ を求める道具です。

## 例:検査の陽性的中率

ある病気の有病率は 1%。検査は、病気の人の 99% で陽性になり(感度)、病気でない人でも 5% で陽性になる(偽陽性)。検査で陽性になった人が、本当に病気である確率は?

**式で計算する**:

- 陽性の確率(全確率の定理):$P(\text{陽性}) = 0.01 \times 0.99 + 0.99 \times 0.05 = 0.0099 + 0.0495 = 0.0594$
- ベイズの定理:$P(\text{病気} \mid \text{陽性}) = \dfrac{0.0099}{0.0594} \approx 0.167$

**人数に置き換えて考える**:10000 人で考えると、次の図のようになります。

![検査の例を 10000 人で考えた樹形図](images/u04_tree.png)

陽性になるのは、病気の人 99 人と、病気でない人 495 人の合わせて 594 人。そのうち本当に病気なのは 99 人なので、99 ÷ 594 ≈ 0.167 です。

感度 99% の検査でも、陽性の人のうち本当に病気の人は約 17% にとどまります。**病気でない人がはるかに多い**(事前確率が低い)ため、偽陽性の人数の方が多くなるからです。

## まとめ

- 全確率の定理:$P(A) = \sum_i P(B_i)P(A \mid B_i)$(原因ごとに分けて足す)。
- ベイズの定理:$P(B_i \mid A) = P(B_i)P(A \mid B_i) / P(A)$。結果から原因の確率(事後確率)を求める。
- 事前確率が低いと、精度の高い検査でも陽性的中率は低くなる。10000 人などの人数に置き換えると考えやすい。
