# 確率変数と期待値・分散
確率的に値が決まる「確率変数」と、その分布の特徴を表す期待値・分散を学びます。1次変換や和の期待値・分散、共分散と相関係数を手計算できるようにします。

## 学習目標
- 離散型・連続型の確率変数と確率分布を説明できる
- 確率分布表から期待値・分散・標準偏差を計算できる
- $aX + b$ や $X + Y$ の期待値・分散を、公式を使って求められる
- 2つの確率変数の共分散・相関係数を求め、独立との関係を説明できる

## 確率変数と確率分布

試行の結果によって値が決まる変数を **確率変数** といいます。

| 種類 | 値のとり方 | 確率の表し方 | 例 |
|---|---|---|---|
| **離散型** | とびとびの値 | 各値の確率 $P(X = x)$(確率関数) | サイコロの目、不良品の個数 |
| **連続型** | 連続的な値 | 確率密度関数 $f(x)$。確率は **面積** | 身長、待ち時間 |

連続型では、ある1点をとる確率は 0 で、「$a$ から $b$ の間」の確率を密度関数の下の面積で表します(「連続型の確率分布」の単元)。

**例**:サイコロを2個投げたときの目の和 $X$ の確率分布

## 期待値と分散

離散型の確率変数 $X$ が値 $x_1, x_2, \dots$ をそれぞれ確率 $p_1, p_2, \dots$ でとるとき

$$E[X] = \sum x_i p_i, \qquad V[X] = E[(X - E[X])^2] = E[X^2] - (E[X])^2$$

標準偏差は $\sqrt{V[X]}$ です。分散は「2乗の期待値 − 期待値の2乗」で計算すると楽です。

**例**:賞金 1000 円が確率 0.1、100 円が確率 0.3、0 円が確率 0.6 のくじ

| 賞金 $x$ | 1000 | 100 | 0 | 合計 |
|---|---|---|---|---|
| 確率 $p$ | 0.1 | 0.3 | 0.6 | 1 |
| $x p$ | 100 | 30 | 0 | $E[X] = 130$ |
| $x^2 p$ | 100000 | 3000 | 0 | $E[X^2] = 103000$ |

分散は $103000 - 130^2 = 103000 - 16900 = 86100$、標準偏差は約 293 円です。

## 1次変換と和の期待値・分散

| 量 | 期待値 | 分散 |
|---|---|---|
| $aX + b$ | $aE[X] + b$ | $a^2 V[X]$ |
| $X + Y$ | $E[X] + E[Y]$(いつでも成り立つ) | $V[X] + V[Y] + 2\,\mathrm{Cov}[X, Y]$ |
| $X - Y$ | $E[X] - E[Y]$ | $V[X] + V[Y] - 2\,\mathrm{Cov}[X, Y]$ |

- **共分散** $\mathrm{Cov}[X, Y] = E[(X - E[X])(Y - E[Y])] = E[XY] - E[X]E[Y]$
- **相関係数** $\rho = \dfrac{\mathrm{Cov}[X, Y]}{\sqrt{V[X]}\sqrt{V[Y]}}$($-1 \le \rho \le 1$)
- $X$ と $Y$ が **独立なら共分散は 0** で、$V[X \pm Y] = V[X] + V[Y]$(差でも分散は足し算)
- 共分散が 0 でも、独立とは限らない(曲線的な関係など)

**例**:$V[X] = 4$、$V[Y] = 9$、$\mathrm{Cov}[X, Y] = 3$ のとき

- $V[X + Y] = 4 + 9 + 2 \times 3 = 19$、$V[X - Y] = 4 + 9 - 2 \times 3 = 7$
- 相関係数 $\rho = 3 / (2 \times 3) = 0.5$

## まとめ

- 期待値 $E[X] = \sum x p$、分散 $V[X] = E[X^2] - (E[X])^2$。連続型では確率は密度関数の下の面積。
- $E[aX + b] = aE[X] + b$、$V[aX + b] = a^2 V[X]$。
- $E[X + Y] = E[X] + E[Y]$。$V[X \pm Y] = V[X] + V[Y] \pm 2\,\mathrm{Cov}[X, Y]$。独立なら共分散は 0。
- 相関係数 $\rho = \mathrm{Cov}[X, Y]/(\sigma_X \sigma_Y)$。共分散 0 でも独立とは限らない。
