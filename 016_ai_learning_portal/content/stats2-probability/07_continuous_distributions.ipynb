# 連続型の確率分布
連続的な値をとる確率変数の分布として、一様分布・指数分布・正規分布・2変量正規分布を学びます。確率を面積として求め、期待値・分散を計算できるようにします。

## 学習目標
- 連続型の確率変数の確率を、確率密度関数の下の面積として求められる
- 一様分布の確率・期待値・分散を計算できる
- 指数分布の確率を計算し、無記憶性とポアソン分布との関係を説明できる
- 2変量正規分布の相関係数と独立の関係を説明できる

## 確率密度関数

連続型の確率変数 $X$ の分布は **確率密度関数** $f(x)$ で表します。

- $f(x) \geq 0$ で、グラフの下の面積の合計は 1
- $P(a \leq X \leq b)$ は、$a$ から $b$ までのグラフの下の面積
- 1点をとる確率は 0(なので $P(a \le X \le b)$ と $P(a < X < b)$ は等しい)

![一様分布・指数分布の密度と、確率 = 面積](images/u07_continuous.png)

## 一様分布

区間 $[a, b]$ のどこでも同じ確からしさで値をとる分布を **一様分布** $U(a, b)$ といいます。密度は区間内で一定の $1/(b - a)$ です。

$$E[X] = \frac{a + b}{2}, \qquad V[X] = \frac{(b - a)^2}{12}$$

**例**:$U(0, 10)$ では、$P(2 < X < 5) = 3 \times \frac{1}{10} = 0.3$(長方形の面積)、期待値 5、分散 $100/12 \approx 8.33$。

## 指数分布

**指数分布** は、まれな出来事が次に起こるまでの **待ち時間** を表す分布です(次の客が来るまでの時間、機械が故障するまでの時間など)。

$$f(x) = \lambda e^{-\lambda x}\ (x \geq 0), \qquad P(X > x) = e^{-\lambda x}, \qquad E[X] = \frac{1}{\lambda}, \qquad V[X] = \frac{1}{\lambda^2}$$

**例**:客が平均 5 分おき($\lambda = 1/5 = 0.2$)に来るとき、次の客が来るまで 10 分以上かかる確率は $e^{-0.2 \times 10} = e^{-2} \approx 0.135$。

- **ポアソン分布との関係**:単位時間あたり平均 $\lambda$ 回起こる出来事(回数はポアソン分布)の、間隔の時間が指数分布に従う
- **無記憶性**:$P(X > s + t \mid X > s) = P(X > t)$。すでに 10 分待ったとしても、さらに 5 分以上待つ確率は、最初から 5 分以上待つ確率と同じ($e^{-1} \approx 0.368$)

## 正規分布と2変量正規分布

**正規分布** $N(\mu, \sigma^2)$ は、平均 $\mu$ を中心とした左右対称のベル型の分布で、測定誤差や身長など多くの現象に当てはまります。確率は標準化して標準正規分布表から求めます(「標本分布と推定」の講座で詳しく扱います)。

2つの確率変数 $(X, Y)$ の同時分布で、それぞれが正規分布に従い、関係の強さを相関係数 $\rho$ で表すものを **2変量正規分布** といいます。

![2変量正規分布の散布図(相関係数 0・0.6・−0.8)](images/u07_bivariate.png)

- $\rho$ が大きいほど、点は右上がりの直線のまわりに集まる(負なら右下がり)
- **2変量正規分布では、$\rho = 0$(無相関)と独立は同じ**。一般の分布では、無相関でも独立とは限らない

連続型の確率分布について述べた次の記述のうち、**正しいもの** を1つ選べ。

- 指数分布に従う待ち時間では、すでに長く待っているほど、さらに待つ時間は短くなる。
- 連続型の確率変数 X について、P(X = 3) は確率密度関数 f(3) に等しい。
- 一様分布 U(0, 10) の確率密度は、区間内で 1 である。
- 2変量正規分布に従う X と Y の相関係数が 0 なら、X と Y は独立である。

## まとめ

- 連続型では、確率は確率密度関数の下の面積。1点の確率は 0。
- 一様分布 $U(a, b)$:$E = (a+b)/2$、$V = (b-a)^2/12$。
- 指数分布:待ち時間の分布。$P(X > x) = e^{-\lambda x}$、$E = 1/\lambda$、$V = 1/\lambda^2$。無記憶性があり、回数がポアソン分布の出来事の間隔。
- 2変量正規分布では、無相関と独立が同じ。
