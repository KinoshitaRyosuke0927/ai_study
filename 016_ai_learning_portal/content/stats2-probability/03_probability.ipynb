# 確率の基礎
推測統計の土台となる確率の基本を学びます。事象の表し方、加法定理と余事象、条件付き確率と乗法定理、事象の独立を、手計算で使えるようにします。

## 学習目標
- 標本空間・事象・和事象・積事象・余事象を説明できる
- 加法定理と余事象を使って確率を計算できる
- 条件付き確率を定義どおりに計算し、乗法定理を使える
- 2つの事象が独立かどうかを判定できる

## 事象と確率

起こりうる結果全体の集合を **標本空間**、その一部を **事象** といいます。すべての結果が同じ程度に起こりやすいとき、事象 $A$ の確率は

$$P(A) = \frac{A \text{ に含まれる結果の数}}{\text{全体の結果の数}}$$

例:サイコロを2個投げて目の和が 7 になる確率は、36 通りのうち (1,6), (2,5), …, (6,1) の 6 通りなので $6/36 = 1/6$。

![事象 A・B のベン図](images/u03_venn.png)

| 事象 | 記号 | 意味 |
|---|---|---|
| 和事象 | $A \cup B$ | A または B が起こる |
| 積事象 | $A \cap B$ | A と B がともに起こる |
| 余事象 | $\bar{A}$ | A が起こらない |
| 排反 | $A \cap B = \emptyset$ | A と B は同時に起こらない |

## 加法定理と余事象

**加法定理**:

$$P(A \cup B) = P(A) + P(B) - P(A \cap B)$$

重なり(積事象)を2回数えないように引きます。A と B が排反なら $P(A \cup B) = P(A) + P(B)$。

**例**:トランプ 52 枚から1枚引くとき、「ハートまたは絵札(J・Q・K)」の確率は

$$\frac{13}{52} + \frac{12}{52} - \frac{3}{52} = \frac{22}{52} = \frac{11}{26}$$

**余事象**:$P(\bar{A}) = 1 - P(A)$。「少なくとも1回」の確率は、余事象「1回も起こらない」を使うと計算しやすくなります。

**例**:サイコロを 3 回投げて、6 の目が少なくとも1回出る確率は $1 - (5/6)^3 \approx 0.421$。

## 条件付き確率と乗法定理

事象 $B$ が起こったという条件のもとで $A$ が起こる確率を **条件付き確率** といいます。

$$P(A \mid B) = \frac{P(A \cap B)}{P(B)}$$

「B が起こった場合だけに絞って、その中で A の割合を見る」と考えます。これを変形したものが **乗法定理** です。

$$P(A \cap B) = P(B)\,P(A \mid B) = P(A)\,P(B \mid A)$$

**例**(「記述統計」の講座の試食会のクロス表):参加者 100 人のうち男性 40 人、男性で購入した人 12 人、購入した人は全体で 39 人。

- $P(\text{購入} \mid \text{男性}) = 12 / 40 = 0.30$(男性に絞った購入率)
- $P(\text{男性} \mid \text{購入}) = 12 / 39 \approx 0.31$(購入者に絞った男性の割合)

$P(A \mid B)$ と $P(B \mid A)$ は一般に異なります。

**乗法定理の例**:赤玉 3 個・白玉 2 個の袋から、戻さずに 2 個取り出すとき、2 個とも赤である確率は $\dfrac{3}{5} \times \dfrac{2}{4} = \dfrac{3}{10}$。

## 事象の独立

一方が起こるかどうかが、もう一方の起こる確率に影響しないとき、2つの事象は **独立** であるといいます。

$$P(A \cap B) = P(A)\,P(B) \quad\Longleftrightarrow\quad P(A \mid B) = P(A)$$

- コインを2回投げるときの1回目と2回目の結果は独立
- 戻さずに取り出すときの1個目と2個目は独立ではない(1個目によって2個目の確率が変わる)
- **排反と独立は別の概念**。$P(A) > 0$, $P(B) > 0$ で排反なら、一方が起こるともう一方は起こらないので、独立ではない

## まとめ

- 加法定理 $P(A \cup B) = P(A) + P(B) - P(A \cap B)$。「少なくとも1回」は余事象 $1 - P(\text{1回も起こらない})$ で求める。
- 条件付き確率 $P(A \mid B) = P(A \cap B)/P(B)$。$P(A \mid B)$ と $P(B \mid A)$ は別物。
- 乗法定理 $P(A \cap B) = P(B)P(A \mid B)$。非復元抽出の確率などに使う。
- 独立 ⇔ $P(A \cap B) = P(A)P(B)$。排反と独立は別の概念。
