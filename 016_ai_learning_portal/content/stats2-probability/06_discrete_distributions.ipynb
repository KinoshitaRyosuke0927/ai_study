# 離散型の確率分布
回数や個数を表す代表的な離散型の確率分布(ベルヌーイ・二項・ポアソン・幾何・超幾何・負の二項)を学びます。場面に合った分布を選び、確率・期待値・分散を手計算できるようにします。

## 学習目標
- 二項分布の確率・期待値・分散を計算できる
- ポアソン分布の確率を計算し、二項分布の近似として使える
- 幾何分布・超幾何分布・負の二項分布の意味を説明し、場面に当てはめられる
- 場面から、適切な確率分布を選べる

## ベルヌーイ分布と二項分布

成功(確率 $p$)か失敗(確率 $1 - p$)かの試行を **ベルヌーイ試行** といいます。1回の試行で成功なら 1、失敗なら 0 をとる確率変数は、期待値 $p$、分散 $p(1-p)$ のベルヌーイ分布に従います。

独立なベルヌーイ試行を $n$ 回くり返したときの成功回数 $X$ は、**二項分布** $B(n, p)$ に従います。

$$P(X = k) = {}_n\mathrm{C}_k\, p^k (1 - p)^{n - k}, \qquad E[X] = np, \qquad V[X] = np(1 - p)$$

**例**:不良率 30% の製品を 10 個調べたとき、不良品がちょうど 2 個である確率は

$${}_{10}\mathrm{C}_2 \times 0.3^2 \times 0.7^8 = 45 \times 0.09 \times 0.0576 \approx 0.234$$

期待値は $10 \times 0.3 = 3$ 個、分散は $10 \times 0.3 \times 0.7 = 2.1$ です。

## ポアソン分布

一定の時間・範囲の中で、まれな出来事が起こる回数は **ポアソン分布** で表されます(1 日の事故件数、1 時間の問い合わせ件数など)。平均を $\lambda$ とすると

$$P(X = k) = \frac{e^{-\lambda}\lambda^k}{k!}, \qquad E[X] = V[X] = \lambda$$

**平均と分散が等しい** のが特徴です。

**例**:1 時間あたり平均 3 件の問い合わせがあるとき、1 時間に問い合わせが 0 件である確率は $e^{-3} \approx 0.050$、1 件である確率は $3e^{-3} \approx 0.149$。

**二項分布の近似**:$n$ が大きく $p$ が小さいとき、$B(n, p)$ は平均 $\lambda = np$ のポアソン分布で近似できます。例えば $B(1000, 0.002)$ で $X = 0$ となる確率は 0.1351、ポアソン分布($\lambda = 2$)では $e^{-2} \approx 0.1353$ です。

## 幾何分布・超幾何分布・負の二項分布

| 分布 | 表すもの | 確率・期待値 |
|---|---|---|
| **幾何分布** | 成功確率 $p$ の試行で、**初めて成功するまで** の試行回数 | $P(X = k) = (1-p)^{k-1}p$、$E[X] = 1/p$ |
| **超幾何分布** | $N$ 個中 $M$ 個が当たりの中から、**戻さずに** $n$ 個取り出したときの当たりの数 | $P(X = k) = \dfrac{{}_M\mathrm{C}_k \cdot {}_{N-M}\mathrm{C}_{n-k}}{{}_N\mathrm{C}_n}$、$E[X] = n\dfrac{M}{N}$ |
| **負の二項分布** | **$r$ 回目の成功** までに必要な試行回数(または失敗回数) | 試行回数の期待値 $r/p$ |

**幾何分布の例**:サイコロで初めて 6 が出るまでの回数の期待値は $1/(1/6) = 6$ 回。3 回目で初めて 6 が出る確率は $(5/6)^2 \times 1/6 \approx 0.116$。

**超幾何分布の例**:20 個中 5 個が不良品の箱から、戻さずに 4 個取り出すとき、不良品の個数の期待値は $4 \times 5/20 = 1$ 個。不良品が 1 個も含まれない確率は ${}_{15}\mathrm{C}_4 / {}_{20}\mathrm{C}_4 \approx 0.282$。

二項分布は「戻して取り出す(独立)」、超幾何分布は「戻さずに取り出す」場面に使います。母集団が大きければ、超幾何分布は二項分布で近似できます。

## まとめ

- 二項分布 $B(n, p)$:成功回数。$E = np$、$V = np(1-p)$。
- ポアソン分布:まれな出来事の回数。$E = V = \lambda$。$n$ 大・$p$ 小の二項分布の近似。
- 幾何分布:初めて成功するまでの回数(期待値 $1/p$)。負の二項分布:$r$ 回目の成功まで。
- 超幾何分布:戻さずに取り出したときの当たりの数(期待値 $nM/N$)。戻すなら二項分布。
