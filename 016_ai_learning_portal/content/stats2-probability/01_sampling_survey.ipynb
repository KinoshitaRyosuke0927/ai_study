# 観察研究と標本調査
データをどのように集めるかで、そこから言えることが決まります。観察研究と実験研究の違い、全数調査と標本調査、無作為抽出の意味と代表的な抽出法、標本誤差と偏り(バイアス)を学びます。

## 学習目標
- 観察研究と実験研究の違いと、それぞれから言えること(関連か因果か)を説明できる
- 全数調査と標本調査の長所・短所を説明できる
- 単純無作為抽出・系統抽出・層化抽出・集落抽出・多段抽出を区別し、場面に合わせて選べる
- 標本誤差と偏り(選択の偏り・無回答・回答の偏りなど)を区別できる

## 観察研究と実験研究

| 研究の種類 | 内容 | 例 | わかること |
|---|---|---|---|
| **観察研究** | 対象に手を加えず、ありのままを観察・調査する | 喫煙習慣と健康状態の調査 | 主に **関連**(相関) |
| **実験研究** | 研究者が処理(条件)を割り当て、その効果を調べる | 新薬と偽薬を無作為に割り当てる臨床試験 | **因果関係**(処理の効果) |

観察研究では、見かけの関連の背後に第3の要因(**交絡因子**)があるかもしれないため、因果関係を主張するのは難しくなります。
実験研究で処理を **無作為に割り当てる** と、交絡因子の影響が処理群と対照群に均等に散らばるので、差を処理の効果と考えることができます(次の単元)。

## 全数調査と標本調査

| 調査 | 内容 | 長所 | 短所 |
|---|---|---|---|
| **全数調査** | 母集団のすべてを調べる | 標本誤差がない | 費用・時間がかかる。破壊検査では不可能 |
| **標本調査** | 母集団の一部(標本)を調べて母集団を推測する | 費用・時間を抑えられる。丁寧に調べられる | 標本誤差がある |

国勢調査は全数調査、内閣支持率の世論調査や製品の抜き取り検査は標本調査の例です。

## 無作為抽出と抽出法

標本から母集団を正しく推測するには、母集団のどの個体も **同じ確率で選ばれる**(偏りなく選ばれる)ようにします。これを **無作為抽出** といいます。

![4つの抽出法のイメージ(赤が選ばれた個体)](images/u01_sampling_methods.png)

| 抽出法 | 方法 | 特徴 |
|---|---|---|
| **単純無作為抽出** | 母集団から乱数などで直接選ぶ | 最も基本。名簿(抽出枠)が必要 |
| **系統抽出** | 最初の1つを無作為に選び、以後一定間隔 $k$ ごとに選ぶ | 手間が少ない。名簿に周期性があると偏る |
| **層化抽出** | 母集団をいくつかの層(年代・地域など)に分け、各層から無作為に選ぶ | 層ごとの違いが大きいとき、推定の精度が上がる |
| **集落(クラスター)抽出** | 母集団を集落(学校・町など)に分け、選んだ集落の全員を調べる | 調査の手間が少ない。集落内が似ていると精度が下がる |
| **多段抽出** | 市町村 → 地区 → 世帯のように、段階的に無作為に選ぶ | 全国調査などで使われる |

**系統抽出の間隔**:母集団 2000 人から 100 人を選ぶなら、間隔は $k = 2000 / 100 = 20$。最初の 1 人を 1〜20 番から無作為に選び、以後 20 人おきに選びます。

**層化抽出の比例配分**:各層の大きさに比例して標本を割り当てます。例えば層の大きさが 4000・3500・2500 人で標本 200 人なら、$200 \times \frac{4000}{10000} = 80$、70、50 人です。

## 標本誤差と偏り

標本調査の結果と本当の値(母数)とのずれには、2種類あります。

| 種類 | 原因 | 減らし方 |
|---|---|---|
| **標本誤差** | 母集団の一部だけを調べることによる、偶然のずれ | 標本を大きくする(無作為抽出なら、推定の精度を計算できる) |
| **偏り(バイアス)** | 調べ方の欠点による、一方向へのずれ | 標本を大きくしても減らない。調査の設計で防ぐ |

偏りの主な原因:

- **選択の偏り**:母集団の一部が選ばれにくい(例:平日昼間の電話調査では、働いている人が選ばれにくい)
- **自己選択の偏り**:回答したい人だけが答える(例:ウェブの任意アンケート)
- **無回答の偏り**:回答しなかった人の意見が、回答した人と異なる
- **回答の偏り**:質問の仕方や、本音を言いにくい質問による偏り(誘導的な質問など)

**大きな標本でも、偏りのある方法で集めたデータからは正しい推測はできません**。

## まとめ

- 観察研究からは主に関連、処理を無作為に割り当てる実験研究からは因果関係がわかる。
- 標本調査では、どの個体も同じ確率で選ばれる無作為抽出が前提。単純無作為・系統・層化・集落・多段抽出を目的に合わせて使う。
- 系統抽出の間隔は N/n、層化抽出の比例配分は n × (層の大きさ / N)。
- 標本誤差は標本を大きくすれば減るが、偏り(選択・自己選択・無回答・回答の偏り)は減らない。調査の設計で防ぐ。
