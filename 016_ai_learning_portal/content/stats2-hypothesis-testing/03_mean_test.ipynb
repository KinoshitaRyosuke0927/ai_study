# 母平均の検定
1つの母集団の平均が、ある値と異なるかを検定します。母分散がわかっているときの z 検定と、わからないときの t 検定を、手計算と統計ソフトの出力の読み取りで身につけます。

## 学習目標
- 母分散が既知のときの z 検定を手計算できる
- 母分散が未知のときの t 検定(自由度 n − 1)を、t 分布表のパーセント点を使って行える
- z 検定と t 検定の使い分けと、t 分布の特徴を説明できる
- 統計ソフトの t 検定の出力から、検定統計量・自由度・p 値を読み取って結論を出せる

## 母分散が既知の場合:z 検定

母標準偏差 $\sigma$ がわかっている正規母集団から大きさ $n$ の標本をとると、標本平均 $\bar{X}$ は平均 $\mu$、標準偏差 $\sigma/\sqrt{n}$(**標準誤差**)の正規分布に従います。
そこで、帰無仮説 $H_0: \mu = \mu_0$ のもとで次の検定統計量は標準正規分布に従います。

$$Z = \frac{\bar{X} - \mu_0}{\sigma / \sqrt{n}}$$

**例**:内容量 500 g と表示された商品を 16 個調べたら、平均は 502.2 g だった。母標準偏差は 4 g とわかっている。表示と異なるといえるか(両側検定、有意水準 5%)。

1. $H_0: \mu = 500$、$H_1: \mu \neq 500$
2. 標準誤差 $4 / \sqrt{16} = 1$、検定統計量 $z = (502.2 - 500) / 1 = 2.2$
3. $|z| = 2.2 > 1.96$ なので、有意水準 5% で $H_0$ を棄却する(p 値は約 0.028)
4. 結論:平均内容量は 500 g と異なる(多く入っている)と判断できる

## 母分散が未知の場合:t 検定

実際には母標準偏差がわからないことがほとんどです。その場合は、標本から求めた不偏分散の平方根 $s$ で代用します。

$$t = \frac{\bar{X} - \mu_0}{s / \sqrt{n}}$$

$\sigma$ を $s$ に置き換えた分だけばらつきが大きくなるので、この統計量は標準正規分布ではなく **自由度 $n - 1$ の t 分布** に従います。

![t 分布と標準正規分布](images/u03_t_vs_normal.png)

- t 分布は 0 を中心に左右対称で、標準正規分布より **裾が重い**(極端な値が出やすい)
- 自由度が大きくなるほど標準正規分布に近づく(自由度が大きければ z のパーセント点で近似できる)
- そのため、t 分布のパーセント点は標準正規分布のもの(1.96 など)より大きい

t 分布の主なパーセント点 $t_\alpha(\text{自由度})$:

| 自由度 | 上側 5% | 上側 2.5% | 上側 0.5% |
|---|---|---|---|
| 4 | 2.132 | 2.776 | 4.604 |
| 9 | 1.833 | 2.262 | 3.250 |
| 15 | 1.753 | 2.131 | 2.947 |
| 19 | 1.729 | 2.093 | 2.861 |
| ∞(標準正規分布) | 1.645 | 1.960 | 2.576 |

**例**:ある部品の長さの母平均が 12 mm かどうかを調べるため 10 個を測ったところ、平均 12.6 mm、標準偏差 $s = 1.2$ mm だった(両側検定、有意水準 5%)。

1. $H_0: \mu = 12$、$H_1: \mu \neq 12$
2. 標準誤差 $1.2 / \sqrt{10} \approx 0.379$、検定統計量 $t = 0.6 / 0.379 \approx 1.58$
3. 自由度 9 の上側 2.5% 点は 2.262。$|t| = 1.58 < 2.262$ なので、$H_0$ は棄却されない

![自由度 9 の t 分布の棄却域と t = 1.581](images/u03_t_test.png)

4. 結論:有意水準 5% では、母平均が 12 mm と異なるとはいえない

**z 検定と t 検定の使い分け**

| 条件 | 検定統計量の分布 |
|---|---|
| 母分散 $\sigma^2$ が既知 | 標準正規分布(z 検定) |
| 母分散が未知($s$ で代用) | 自由度 $n - 1$ の t 分布(t 検定) |
| 母分散が未知でも $n$ が十分大きい | 標準正規分布で近似してよい |

## 統計ソフトの出力を読む

上の部品の例を統計ソフトで検定すると、次のような出力が得られます。

```
One-sample t-test   (H0: mean = 12)
  n = 10   mean = 12.600   sd = 1.200
  t = 1.581   df = 9   p-value (two-sided) = 0.148
```

| 項目 | 意味 |
|---|---|
| `t` | 検定統計量 |
| `df` | 自由度($n - 1$) |
| `p-value (two-sided)` | 両側検定の p 値。片側検定の p 値は(向きが合っていれば)この半分 |

p 値 0.148 は 0.05 より大きいので、有意水準 5% では棄却されません。手計算の結論と一致します。

## まとめ

- 母分散が既知なら $Z = (\bar{X} - \mu_0)/(\sigma/\sqrt{n})$ を標準正規分布で検定する(z 検定)。
- 母分散が未知なら $t = (\bar{X} - \mu_0)/(s/\sqrt{n})$ を自由度 $n - 1$ の t 分布で検定する(t 検定)。t 分布は裾が重く、パーセント点は z より大きい。
- 同じデータでも、両側か片側かで結論が変わることがある。片側の p 値は両側の半分(向きが合っている場合)。
- 統計ソフトの出力では、t(検定統計量)・df(自由度)・p-value を読み、p 値と有意水準を比べる。
