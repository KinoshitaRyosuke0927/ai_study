# 2つの母平均の差の検定
2つのグループの平均に差があるかを検定します。母分散が既知の場合、未知で等しいと仮定できる場合、等しいとは限らない場合(ウェルチの検定)、対応のあるデータの場合の4つを、手計算で使い分けられるようにします。

## 学習目標
- 母分散が既知のときの2標本の z 検定を手計算できる
- 等分散を仮定するときのプールした分散を求め、自由度 $n_1 + n_2 - 2$ の t 検定を行える
- 等分散を仮定できないときにウェルチの検定を使うことを説明できる
- 対応のあるデータを、差をとって1標本の t 検定として扱える
- 状況に合わせて、4つの検定を使い分けられる

## 母分散が既知の場合

2つの独立な正規母集団から、それぞれ大きさ $n_1$, $n_2$ の標本をとります。標本平均の差 $\bar{X}_1 - \bar{X}_2$ の分散は、それぞれの分散の **和** $\sigma_1^2/n_1 + \sigma_2^2/n_2$ になります(差の分散なのに足し算になる点に注意)。

帰無仮説 $H_0: \mu_1 = \mu_2$ のもとで

$$Z = \frac{\bar{X}_1 - \bar{X}_2}{\sqrt{\dfrac{\sigma_1^2}{n_1} + \dfrac{\sigma_2^2}{n_2}}}$$

は標準正規分布に従います。

**例**:A 校・B 校の各 25 人の平均点が 78 点・74 点、母分散が 36・64 とわかっているとき、標準誤差は $\sqrt{36/25 + 64/25} = \sqrt{4} = 2$、$z = (78 - 74)/2 = 2.0$。両側 5% の 1.96 を超えるので、差があると判断します。

## 母分散が未知で、等しいと仮定できる場合

2つの母分散が等しい($\sigma_1^2 = \sigma_2^2$)と仮定できるときは、2つの標本の不偏分散 $s_1^2$, $s_2^2$ を自由度で重みづけして平均した **プールした分散**(合併した分散)を使います。

$$s_p^2 = \frac{(n_1 - 1)s_1^2 + (n_2 - 1)s_2^2}{n_1 + n_2 - 2}$$

$$t = \frac{\bar{X}_1 - \bar{X}_2}{\sqrt{s_p^2\left(\dfrac{1}{n_1} + \dfrac{1}{n_2}\right)}} \quad \text{(自由度 } n_1 + n_2 - 2 \text{ の t 分布)}$$

**例**:新しい教え方のクラス(10 人、平均 75 点、不偏分散 36)と従来のクラス(12 人、平均 70 点、不偏分散 40)で、平均に差があるか(両側検定、有意水準 5%)。

1. プールした分散 $s_p^2 = (9 \times 36 + 11 \times 40) / 20 = 764 / 20 = 38.2$
2. 標準誤差 $\sqrt{38.2 \times (1/10 + 1/12)} = \sqrt{7.003} \approx 2.646$
3. $t = (75 - 70) / 2.646 \approx 1.89$。自由度 20 の上側 2.5% 点は 2.086 なので、棄却しない
4. 結論:有意水準 5% では、2つの教え方の平均点に差があるとはいえない(p 値は約 0.07)

## 母分散が未知で、等しいとは限らない場合:ウェルチの検定

2つの母分散が等しいと仮定できないときは、それぞれの不偏分散をそのまま使います。

$$t = \frac{\bar{X}_1 - \bar{X}_2}{\sqrt{\dfrac{s_1^2}{n_1} + \dfrac{s_2^2}{n_2}}}$$

この統計量は近似的に t 分布に従い、自由度は次の **ウェルチ・サタースウェイトの式** で求めます(整数にならないことが多い)。

$$\nu \approx \frac{\left(\dfrac{s_1^2}{n_1} + \dfrac{s_2^2}{n_2}\right)^2}{\dfrac{(s_1^2/n_1)^2}{n_1 - 1} + \dfrac{(s_2^2/n_2)^2}{n_2 - 1}}$$

上の教え方の例にウェルチの検定を使うと、$t \approx 1.90$、自由度 $\nu \approx 19.6$ で、結論は変わりません。分散の比が大きく異なるときほど、等分散を仮定した t 検定との差が大きくなります。

## 対応のあるデータ

同じ人の「前」と「後」のように、2つのデータが **ペアになっている** ときは、ペアごとの差 $d_i$ をとり、差の平均が 0 かどうかを1標本の t 検定で調べます。

$$t = \frac{\bar{d}}{s_d / \sqrt{n}} \quad \text{(自由度 } n - 1 \text{ の t 分布)}$$

![対応のあるデータと、差だけを見る考え方](images/u05_paired.png)

**例**:5 人の講習前後の点数の差(後 − 前)が `3, 5, 2, 6, 4` だった。

1. 差の平均 $\bar{d} = 20 / 5 = 4$、偏差平方和 $1 + 1 + 4 + 4 + 0 = 10$、不偏分散 $s_d^2 = 10 / 4 = 2.5$
2. 標準誤差 $\sqrt{2.5} / \sqrt{5} \approx 0.707$、$t = 4 / 0.707 \approx 5.66$
3. 自由度 4 の上側 2.5% 点は 2.776 なので、両側 5% で棄却する。講習で点数は上がったと判断できる

図のように、人によって点数の水準は大きく違いますが、全員が少しずつ上がっています。2つのグループとして比べると個人差に埋もれてしまう差も、ペアの差を見ると検出しやすくなります。

## 検定の使い分け

| 状況 | 検定統計量 | 分布(自由度) |
|---|---|---|
| 独立な2群、母分散が既知 | $Z$(分母は $\sqrt{\sigma_1^2/n_1 + \sigma_2^2/n_2}$) | 標準正規分布 |
| 独立な2群、母分散は未知だが等しい | プールした分散を使う $t$ | t 分布($n_1 + n_2 - 2$) |
| 独立な2群、母分散が等しいとは限らない | ウェルチの $t$ | t 分布(ウェルチの式による自由度) |
| 対応のあるデータ(同じ対象の前後など) | 差の $t$ | t 分布($n - 1$) |

## まとめ

- 母分散が既知なら $Z = (\bar{X}_1 - \bar{X}_2)/\sqrt{\sigma_1^2/n_1 + \sigma_2^2/n_2}$(差の分散は和)。
- 等分散を仮定できるなら、プールした分散 $s_p^2$ を使い自由度 $n_1 + n_2 - 2$ の t 検定。仮定できないならウェルチの検定(自由度はウェルチの式)。
- 対応のあるデータは、ペアの差をとって自由度 $n - 1$ の1標本 t 検定にする。個人差の影響を除ける。
- 統計ソフトの出力では、検定の種類・t・df・p 値を確認し、t の符号から差の向きを読む。
