# 独立性の検定
2つの質的変数に関連があるかを、2元クロス表とカイ二乗値で検定します。期待度数・カイ二乗値・自由度を手計算し、統計ソフトの出力を読み取れるようにします。

## 学習目標
- 独立性の帰無仮説のもとで、クロス表の各マスの期待度数を求められる
- 独立性の検定のカイ二乗値と自由度 $(r-1)(c-1)$ を手計算できる
- 2 × 2 表のイェーツの補正の考え方を説明できる
- 統計ソフトの出力から結論を出し、p 値と関連の強さの違いを説明できる

## 独立性の検定の考え方

「記述統計」の講座で、試食会の参加者 100 人の性別と購入の有無をまとめた次のクロス表を見ました。

| | 購入あり | 購入なし | 計 |
|---|---|---|---|
| 男性 | 12 | 28 | 40 |
| 女性 | 27 | 33 | 60 |
| 計 | 39 | 61 | 100 |

購入率は男性 30%、女性 45% と違いますが、この差は偶然の範囲でしょうか。

帰無仮説 $H_0$:**性別と購入の有無は独立である**(関連がない)

独立なら、男女どちらも全体の購入率 39% で購入するはずです。

![性別ごとの購入の構成比と、独立なときの構成比](images/u08_band.png)

## 期待度数とカイ二乗値

独立なときの各マスの期待度数は、周辺度数から次のように求めます。

$$E_{ij} = \frac{(\text{行 } i \text{ の合計}) \times (\text{列 } j \text{ の合計})}{\text{総数}}$$

| | 購入あり | 購入なし |
|---|---|---|
| 男性 | 40 × 39 ÷ 100 = 15.6 | 40 × 61 ÷ 100 = 24.4 |
| 女性 | 60 × 39 ÷ 100 = 23.4 | 60 × 61 ÷ 100 = 36.6 |

カイ二乗値は、すべてのマスについて $(O - E)^2/E$ を合計します。

$$\chi^2 = \frac{(12 - 15.6)^2}{15.6} + \frac{(28 - 24.4)^2}{24.4} + \frac{(27 - 23.4)^2}{23.4} + \frac{(33 - 36.6)^2}{36.6} \approx 0.831 + 0.531 + 0.554 + 0.354 = 2.27$$

2 × 2 表では、どのマスも $|O - E|$ が同じ(ここでは 3.6)になる点も確かめておきましょう。

## 自由度と判断

独立性の検定の自由度は、$r$ 行 $c$ 列の表なら

$$(r - 1)(c - 1)$$

です。周辺度数が決まっていると、$(r-1)(c-1)$ 個のマスの度数を決めれば残りが決まるためです。2 × 2 表なら自由度は 1、3 × 2 表なら 2 です。

例の表は自由度 1 で、上側 5% 点は 3.841 です。$\chi^2 = 2.27 < 3.841$ なので、有意水準 5% では帰無仮説を棄却しません(p 値は約 0.13)。

![自由度 1 のカイ二乗分布の棄却域と χ² = 2.27](images/u08_chi2_df1.png)

結論:有意水準 5% では、性別と購入の有無に関連があるとはいえません。

**イェーツの補正**:2 × 2 表では、各マスの $|O - E|$ から 0.5 を引いてから2乗する補正(連続修正)をすることがあります。補正するとカイ二乗値は小さくなり(この例では約 1.68)、棄却されにくくなります。統計ソフトでは補正ありが既定のこともあるので、出力を見るときは確認します。

適合度検定と同じく、期待度数が小さい(5 未満)マスが多いときは、カイ二乗分布による近似が悪くなる点に注意します。

## p 値と関連の強さ

独立性の検定でわかるのは「関連がないとはいえない(関連がある)かどうか」で、**関連の強さ** ではありません。
総数が大きいと、わずかな差でもカイ二乗値は大きくなり、p 値は小さくなります。関連の強さは、クロス表の行パーセント(購入率の差など)を見て判断します。

## まとめ

- 独立性の検定の帰無仮説は「2つの変数は独立(関連がない)」。期待度数は(行の合計 × 列の合計)÷ 総数。
- カイ二乗値 $\sum (O-E)^2/E$ を、自由度 $(r-1)(c-1)$ のカイ二乗分布の上側パーセント点と比べる。
- 2 × 2 表ではイェーツの補正をすることがあり、補正するとカイ二乗値は小さくなる。
- 検定でわかるのは関連の有無であり、関連の強さは行パーセントなどで判断する。
