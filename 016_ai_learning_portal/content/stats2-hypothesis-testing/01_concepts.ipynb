# 仮説検定の考え方
「偶然では説明しにくい差か」をデータから判断する仮説検定の手順を学びます。帰無仮説と対立仮説、検定統計量、有意水準と棄却域、p 値の意味を、手計算の例で身につけます。

## 学習目標
- 帰無仮説・対立仮説を立て、両側検定と片側検定を使い分けられる
- 検定統計量を計算し、棄却域または p 値から結論を導ける
- p 値と有意水準の意味を正しく説明できる
- 「帰無仮説が棄却されない」ことの正しい解釈を説明できる

## 仮説検定とは

コインを 100 回投げたら、表が 60 回出ました。このコインは公平(表の確率が 0.5)だといえるでしょうか。
公平なコインでも、表がちょうど 50 回になるとは限りません。問題は、**60 回という結果が「偶然のばらつき」で説明できる範囲かどうか** です。

仮説検定は、次のような **背理法** に似た考え方で判断します。

1. 「コインは公平である」と仮定する(**帰無仮説**)
2. その仮定のもとで、観測された結果(またはそれ以上に極端な結果)が起こる確率を計算する
3. その確率がとても小さければ、「仮定がおかしい」と考えて仮定を捨てる(**棄却する**)

## 帰無仮説と対立仮説

| 仮説 | 記号 | 内容 | コインの例 |
|---|---|---|---|
| **帰無仮説** | $H_0$ | 「差がない」「効果がない」という、否定したい仮説 | $p = 0.5$ |
| **対立仮説** | $H_1$ | 「差がある」という、示したい仮説 | $p \neq 0.5$ |

対立仮説の立て方で、検定は次の2種類に分かれます。

| 種類 | 対立仮説の例 | 使う場面 |
|---|---|---|
| **両側検定** | $p \neq 0.5$ | 大きい・小さいのどちらの向きの差も知りたい |
| **片側検定** | $p > 0.5$(右片側)、$p < 0.5$(左片側) | 向きがはっきり決まっている(「改善したか」「減ったか」など) |

片側か両側かは、**データを見る前に、調べたいことから決めます**。データを見てから都合のよい方を選ぶのは誤りです。

## 検定統計量と有意水準・棄却域

データから計算し、帰無仮説が正しいときの分布がわかっている量を **検定統計量** といいます。コインの例では、表の回数 $X$ を標準化した値を使います。

帰無仮説 $p = 0.5$ のもとで、$n = 100$ 回のうち表の回数は平均 $np = 50$、標準偏差 $\sqrt{np(1-p)} = \sqrt{25} = 5$ で、近似的に正規分布に従います。そこで

$$z = \frac{X - np}{\sqrt{np(1-p)}} = \frac{60 - 50}{5} = 2.0$$

帰無仮説が正しければ、$z$ はおよそ標準正規分布に従います。

**有意水準** $\alpha$ は、「この確率より起こりにくいことが起きたら帰無仮説を棄却する」という基準で、ふつう 5%(0.05)や 1%(0.01)を使います。
有意水準に対応して、帰無仮説を棄却する $z$ の範囲を **棄却域** といいます。

![標準正規分布の棄却域](images/u01_regions.png)

標準正規分布の主なパーセント点(上側確率がその値になる点):

| 上側確率 | 0.05 | 0.025 | 0.01 | 0.005 |
|---|---|---|---|---|
| パーセント点 | 1.645 | 1.96 | 2.326 | 2.576 |

- 両側検定・有意水準 5%:$|z| > 1.96$ なら棄却(両側に 2.5% ずつ)
- 右片側検定・有意水準 5%:$z > 1.645$ なら棄却
- 両側検定・有意水準 1%:$|z| > 2.576$ なら棄却

コインの例(両側検定)では $|z| = 2.0 > 1.96$ なので、有意水準 5% で帰無仮説を棄却し、「コインは公平でない」と判断します。有意水準 1% なら $2.0 < 2.576$ なので棄却しません。

## p 値

**p 値** は、帰無仮説が正しいと仮定したとき、**観測された結果、またはそれより極端な結果が得られる確率** です。

![z = 2.0 の両側 p 値](images/u01_pvalue.png)

コインの例では、標準正規分布で $z \geq 2.0$ となる確率は 0.0228 です。両側検定なので反対側も合わせて、p 値は $0.0228 \times 2 = 0.0455$ になります。

判断のルール:**p 値 < 有意水準 なら帰無仮説を棄却する**。
p 値 0.0455 は 0.05 より小さいので 5% では棄却、0.01 より大きいので 1% では棄却しません。棄却域で判断した結果と必ず一致します。

注意:この例は正規近似で計算しています。二項分布で正確に計算する(または連続修正をする)と、p 値は約 0.057 になり、5% では棄却されません。p 値が有意水準に近い「境界ぎりぎり」の場合は、近似の仕方で結論が変わることがあります。

## 結論の述べ方

| 結果 | 結論の述べ方 | よくある誤り |
|---|---|---|
| 棄却された | 「有意水準 5% で、公平でないと判断できる(有意な差がある)」 | 「100% 公平でない」と断定する |
| 棄却されなかった | 「有意水準 5% では、公平でないとはいえない」 | 「公平であることが証明された」 |

**帰無仮説が棄却されないことは、帰無仮説が正しいことの証明にはなりません**。データが少なくて差を検出できなかっただけかもしれないからです(次の単元の「検出力」)。

また、p 値は「帰無仮説が正しい確率」ではありません。帰無仮説が正しいと **仮定したうえで** 計算した、データの起こりにくさを表す値です。

**仮説検定の手順のまとめ**

1. 帰無仮説 $H_0$ と対立仮説 $H_1$ を立てる(両側か片側かを決める)
2. 有意水準 $\alpha$ を決める
3. 検定統計量を計算する
4. 棄却域に入るか(または p 値 < $\alpha$ か)を調べる
5. 結論を述べる

## まとめ

- 帰無仮説 $H_0$(差がない)を仮定し、データの起こりにくさを調べて、起こりにくければ $H_0$ を棄却する。
- 両側か片側かは、調べたいことからデータを見る前に決める。
- 検定統計量が棄却域に入る、または p 値 < 有意水準 なら棄却。標準正規分布の上側 5%・2.5%・1%・0.5% 点は 1.645・1.96・2.326・2.576。
- p 値は「帰無仮説のもとで観測結果以上に極端な結果が出る確率」であり、帰無仮説が正しい確率ではない。棄却されないことは帰無仮説の証明ではない。
