# 母比率の検定と母分散の検定
支持率や不良率のような「割合」が、ある値と異なるかを調べる母比率の検定と、ばらつきの大きさが基準と異なるかを調べる母分散の検定(カイ二乗分布)を学びます。

## 学習目標
- 母比率の検定統計量を、正規近似を使って手計算できる
- 母分散の検定統計量 $(n-1)s^2/\sigma_0^2$ を計算し、カイ二乗分布のパーセント点で判断できる
- カイ二乗分布の特徴(正の値・右に裾が長い・平均は自由度)を説明できる
- 検定の種類と、検定統計量が従う分布を対応づけられる

## 母比率の検定

成功(はい・不良など)の割合が $p$ の母集団から大きさ $n$ の標本をとり、成功の割合(標本比率)が $\hat{p}$ だったとします。
$n$ が大きければ、$\hat{p}$ は近似的に平均 $p$、標準偏差 $\sqrt{p(1-p)/n}$ の正規分布に従います(二項分布の正規近似)。

帰無仮説 $H_0: p = p_0$ のもとで、次の検定統計量は近似的に標準正規分布に従います。

$$Z = \frac{\hat{p} - p_0}{\sqrt{p_0(1 - p_0)/n}}$$

標準偏差には、標本比率 $\hat{p}$ ではなく **帰無仮説の値 $p_0$** を使う点に注意します。

**例**:これまでの支持率は 40% だった。600 人に調査したところ、264 人(44%)が支持した。支持率は変わったといえるか(両側検定、有意水準 5%)。

1. $H_0: p = 0.4$、$H_1: p \neq 0.4$
2. 標準偏差 $\sqrt{0.4 \times 0.6 / 600} = \sqrt{0.0004} = 0.02$
3. $z = (0.44 - 0.40) / 0.02 = 2.0$。$|z| > 1.96$ なので棄却する(p 値は約 0.046)
4. 結論:有意水準 5% で、支持率は 40% から変わったと判断できる

## カイ二乗分布

正規母集団から大きさ $n$ の標本をとり、不偏分散を $s^2$ とすると、

$$\chi^2 = \frac{(n-1)s^2}{\sigma^2}$$

は **自由度 $n - 1$ のカイ二乗分布** に従います。

![カイ二乗分布の形](images/u04_chi2_shapes.png)

- 0 以上の値だけをとり、右に裾が長い
- 平均は自由度に等しい(自由度 9 なら平均 9)
- 自由度が大きくなるほど、左右対称に近づく

カイ二乗分布は左右対称ではないので、両側検定では **下側と上側で別々のパーセント点** を使います。

| 自由度 | 下側 2.5% 点(上側 97.5% 点) | 上側 5% 点 | 上側 2.5% 点 |
|---|---|---|---|
| 9 | 2.700 | 16.919 | 19.023 |
| 15 | 6.262 | 24.996 | 27.488 |
| 19 | 8.907 | 30.144 | 32.852 |

## 母分散の検定

帰無仮説 $H_0: \sigma^2 = \sigma_0^2$ のもとで、$\chi^2 = (n-1)s^2 / \sigma_0^2$ は自由度 $n - 1$ のカイ二乗分布に従います。

**例**:ある機械で作る部品の重さの分散は、これまで 4 だった。新しい機械で 10 個作ったところ、不偏分散は $s^2 = 6.0$ だった。分散は変わったといえるか(両側検定、有意水準 5%)。

1. $H_0: \sigma^2 = 4$、$H_1: \sigma^2 \neq 4$
2. $\chi^2 = (10 - 1) \times 6.0 / 4 = 13.5$
3. 自由度 9 の棄却域は $\chi^2 < 2.700$ または $\chi^2 > 19.023$。13.5 はどちらにも入らないので、棄却しない

![自由度 9 のカイ二乗分布の棄却域と χ² = 13.5](images/u04_var_test.png)

4. 結論:有意水準 5% では、分散が変わったとはいえない

ばらつきが「大きくなったか」だけを調べたい(品質の悪化を見つけたい)ときは、右片側検定にして上側 5% 点と比べます。

## まとめ

- 母比率の検定:$Z = (\hat{p} - p_0)/\sqrt{p_0(1-p_0)/n}$(標準偏差には $p_0$ を使う)を標準正規分布で判断する。
- 母分散の検定:$\chi^2 = (n-1)s^2/\sigma_0^2$ を自由度 $n - 1$ のカイ二乗分布で判断する。
- カイ二乗分布は正の値をとり右に裾が長く、平均は自由度。両側検定では下側・上側の2つのパーセント点を使う。
- 母平均(σ 既知)・母比率 → 標準正規分布、母平均(σ 未知)→ t 分布、母分散 → カイ二乗分布。
