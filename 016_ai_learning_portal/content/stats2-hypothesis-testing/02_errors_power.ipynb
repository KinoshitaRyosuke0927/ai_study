# 2種類の過誤と検出力
仮説検定は確率にもとづく判断なので、誤ることがあります。第1種の過誤・第2種の過誤と検出力の関係を図で理解し、検出力を手計算で求められるようにします。

## 学習目標
- 第1種の過誤・第2種の過誤を、具体的な場面に当てはめて説明できる
- 有意水準 α・β・検出力 1 − β の関係を、分布の図で説明できる
- 標本の大きさ・有意水準・真の差の大きさが検出力に与える影響を説明できる
- 簡単な場合の検出力を手計算できる

## 2種類の過誤

検定の判断と、本当の状態の組み合わせは次の4通りです。

| | 帰無仮説を棄却しない | 帰無仮説を棄却する |
|---|---|---|
| **帰無仮説が正しい** | 正しい判断 | **第1種の過誤**(確率 $\alpha$) |
| **帰無仮説が誤り**(対立仮説が正しい) | **第2種の過誤**(確率 $\beta$) | 正しい判断(確率 $1 - \beta$ = **検出力**) |

- **第1種の過誤**:本当は差がないのに「差がある」と判断してしまう誤り(あわてものの誤り)。その確率が **有意水準 $\alpha$** です
- **第2種の過誤**:本当は差があるのに「差があるとはいえない」と判断してしまう誤り(ぼんやりものの誤り)。その確率を $\beta$ で表します
- **検出力** $1 - \beta$:本当に差があるときに、正しく帰無仮説を棄却できる確率

新薬の検定(帰無仮説「効果がない」)なら、第1種の過誤は「効果のない薬を効果があると判断する」、第2種の過誤は「効果のある薬を見逃す」ことにあたります。

## α・β・検出力を図で見る

**例**:ある製品の重さの母平均について、帰無仮説 $H_0: \mu = 50$、対立仮説 $H_1: \mu > 50$ を、有意水準 5% の右片側検定で調べる。母標準偏差は $\sigma = 6$、標本の大きさは $n = 16$ とする。

標本平均の標準偏差(標準誤差)は $\sigma / \sqrt{n} = 6 / 4 = 1.5$ なので、棄却の境界は

$$c = 50 + 1.645 \times 1.5 \approx 52.47$$

標本平均がこれより大きければ $H_0$ を棄却します。ここで、本当の母平均が 53 だったとすると、次の図のようになります。

![H0 と H1 の分布と、α・β・検出力](images/u02_alpha_beta.png)

- 赤い部分(H0 の分布で境界より右):$\alpha = 0.05$
- 紫の部分(H1 の分布で境界より左):$\beta$。本当は $\mu = 53$ なのに棄却できない確率
- 斜線の部分(H1 の分布で境界より右):検出力 $1 - \beta$

β の計算:本当の分布 $N(53, 1.5^2)$ で、標本平均が 52.47 より小さくなる確率です。

$$\beta = P\left(Z < \frac{52.47 - 53}{1.5}\right) = P(Z < -0.355) \approx 0.36, \qquad \text{検出力} = 1 - 0.36 = 0.64$$

## 検出力を高めるには

| 変えること | α | β | 検出力 | 理由 |
|---|---|---|---|---|
| 標本の大きさ $n$ を大きくする | そのまま | 小さくなる | 大きくなる | 標準誤差が小さくなり、2つの分布の重なりが減る |
| 有意水準 $\alpha$ を大きくする(例 1% → 5%) | 大きくなる | 小さくなる | 大きくなる | 棄却の境界が左に動く |
| 真の差(効果)が大きい | そのまま | 小さくなる | 大きくなる | 2つの分布が離れる |

$n$ を 16 から 64 に増やすと、標準誤差は 1.5 から 0.75 に半減し、β は約 0.36 から約 0.01 に下がります。

![標本の大きさによる β の違い](images/u02_sample_size.png)

**α と β はトレードオフ**:$n$ を変えずに有意水準を小さくする(α を 5% → 1%)と、第1種の過誤は減りますが、棄却の境界が右に動くため β は大きくなります(上の例では約 0.36 → 約 0.63)。両方を小さくするには、標本の大きさを増やす必要があります。

2種類の過誤について述べた次の記述のうち、**正しいもの** を1つ選べ。

- 第1種の過誤の確率と第2種の過誤の確率の和は、常に 1 である。
- 検出力は、第1種の過誤の確率 α を 1 から引いたものである。
- 標本の大きさを変えずに有意水準を小さくすると、第1種の過誤の確率は小さくなるが、第2種の過誤の確率は大きくなる。
- 第2種の過誤は、帰無仮説が正しいのに棄却してしまう誤りである。

## まとめ

- 第1種の過誤:$H_0$ が正しいのに棄却する(確率 α = 有意水準)。第2種の過誤:$H_0$ が誤りなのに棄却しない(確率 β)。
- 検出力 $1 - \beta$ は、本当に差があるときに正しく棄却できる確率。真の分布で棄却域に入る確率として計算する。
- 検出力は、標本の大きさ・有意水準・真の差が大きいほど大きい。
- 標本の大きさが同じなら α と β はトレードオフ。両方を小さくするには標本を大きくする。
