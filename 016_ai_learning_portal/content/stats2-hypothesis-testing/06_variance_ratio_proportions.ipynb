# 母分散の比の検定と母比率の差の検定
2つのグループのばらつきが等しいかを調べる F 検定と、2つのグループの割合に差があるかを調べる母比率の差の検定を学びます。

## 学習目標
- 2つの不偏分散の比を F 分布で検定でき、F 分布の自由度を正しく指定できる
- F 分布の下側パーセント点を、上側パーセント点の逆数から求められる
- 母比率の差の検定で、プールした比率を使って検定統計量を手計算できる

## F 分布

2つの独立な正規母集団から大きさ $n_1$, $n_2$ の標本をとり、不偏分散を $s_1^2$, $s_2^2$ とします。2つの母分散が等しいとき

$$F = \frac{s_1^2}{s_2^2}$$

は **自由度 $(n_1 - 1, n_2 - 1)$ の F 分布** に従います。自由度は「分子の自由度, 分母の自由度」の順に書きます。

- 0 以上の値をとり、右に裾が長い。母分散が等しければ、F は 1 の近くに出やすい
- 下側のパーセント点は、自由度を入れ替えた上側のパーセント点の逆数で求められる

$$F_{1-\alpha}(m, n) = \frac{1}{F_{\alpha}(n, m)}$$

例:自由度 (10, 15) の下側 2.5% 点は、$1 / F_{0.025}(15, 10) = 1 / 3.522 \approx 0.284$

F 分布の上側パーセント点の例:

| 自由度 | 上側 5% 点 | 上側 2.5% 点 |
|---|---|---|
| (10, 15) | 2.544 | 3.060 |
| (15, 10) | 2.845 | 3.522 |
| (12, 9) | 3.073 | 3.868 |
| (9, 12) | 2.796 | 3.436 |

## 母分散の比の検定(F 検定)

帰無仮説 $H_0: \sigma_1^2 = \sigma_2^2$ のもとで、$F = s_1^2 / s_2^2$ を自由度 $(n_1 - 1, n_2 - 1)$ の F 分布で判断します。
手計算では、**大きい方の不偏分散を分子** にすると、上側のパーセント点だけで判断できて便利です(両側検定なら、上側 $\alpha/2$ 点と比べる)。

**例**:A 社の部品 11 個の不偏分散は 24、B 社の部品 16 個の不偏分散は 10 だった。ばらつきに差があるか(両側検定、有意水準 5%)。

1. $H_0: \sigma_A^2 = \sigma_B^2$、$H_1: \sigma_A^2 \neq \sigma_B^2$
2. $F = 24 / 10 = 2.4$、自由度は (11 − 1, 16 − 1) = (10, 15)
3. 上側 2.5% 点は 3.060。$F = 2.4 < 3.060$ なので棄却しない

![自由度 (10, 15) の F 分布の棄却域と F = 2.4](images/u06_f_test.png)

4. 結論:有意水準 5% では、2社の部品のばらつきに差があるとはいえない

F 検定は、2つの平均の差の t 検定で「等分散を仮定してよいか」を確かめるときにも使われます。

## 母比率の差の検定

2つのグループの成功の割合 $p_1$, $p_2$ に差があるかを調べます。帰無仮説 $H_0: p_1 = p_2$ のもとでは2つの比率は等しいので、2つの標本を合わせた **プールした比率** で共通の比率を推定します。

$$\hat{p} = \frac{x_1 + x_2}{n_1 + n_2}, \qquad Z = \frac{\hat{p}_1 - \hat{p}_2}{\sqrt{\hat{p}(1 - \hat{p})\left(\dfrac{1}{n_1} + \dfrac{1}{n_2}\right)}}$$

$Z$ は、$n_1$, $n_2$ が大きければ近似的に標準正規分布に従います。

**例**:広告 A を見た 200 人のうち 60 人(30%)、広告 B を見た 200 人のうち 45 人(22.5%)が商品を購入した。購入率に差があるか(両側検定、有意水準 5%)。

1. プールした比率 $\hat{p} = (60 + 45) / 400 = 0.2625$
2. 標準誤差 $\sqrt{0.2625 \times 0.7375 \times (1/200 + 1/200)} \approx 0.0440$
3. $z = (0.300 - 0.225) / 0.0440 \approx 1.70$。$|z| < 1.96$ なので棄却しない(p 値は約 0.09)
4. 結論:有意水準 5% では、2つの広告の購入率に差があるとはいえない

F 検定と母比率の差の検定について述べた次の記述のうち、**誤っているもの** を1つ選べ。

- F 分布の自由度は、(分子の自由度, 分母の自由度)の順に指定する。
- 自由度 (m, n) の F 分布の下側 2.5% 点は、自由度 (n, m) の F 分布の上側 2.5% 点の逆数である。
- F 統計量は、2つの母分散が等しいとき負の値をとることがある。
- 母比率の差の検定では、帰無仮説のもとで、2つの標本を合わせた比率を共通の比率の推定に使う。
- 母比率の差の検定統計量は、標本が大きければ近似的に標準正規分布に従う。

## まとめ

- 母分散の比の検定:$F = s_1^2/s_2^2$ を自由度 $(n_1 - 1, n_2 - 1)$ の F 分布で判断する。大きい方を分子にすると上側パーセント点だけで判断できる。
- F 分布の下側パーセント点は、自由度を入れ替えた上側パーセント点の逆数。
- 母比率の差の検定:プールした比率 $\hat{p} = (x_1 + x_2)/(n_1 + n_2)$ を使った $Z$ を標準正規分布で判断する。
