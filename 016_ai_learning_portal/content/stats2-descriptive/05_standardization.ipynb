# 中心と散らばりの活用:標準化・変動係数・指数化
平均と標準偏差を組み合わせて、単位や水準の違うデータを比べる方法(標準化・偏差値・変動係数・指数化)を学びます。どれも手計算で求められるようにします。

## 学習目標
- データを1次変換したときの平均・分散・標準偏差の変化を説明し、計算できる
- 標準化(z 得点)と偏差値を計算し、異なる科目・集団の値を比較できる
- 変動係数で、平均の大きさが違うデータの散らばりを比べられる
- 基準時点を 100 とする指数を求め、指数の比から変化率を計算できる

## データの変換と平均・標準偏差

データ $x$ を $y = a x + b$ と変換すると、平均・分散・標準偏差は次のように変わります。試験で非常によく使う性質です。

| 指標 | 変換後 | ポイント |
|---|---|---|
| 平均 | $\bar{y} = a\bar{x} + b$ | 平均にも同じ変換をする |
| 分散 | $s_y^2 = a^2 s_x^2$ | 足した $b$ は影響しない。$a$ は2乗で効く |
| 標準偏差 | $s_y = \lvert a \rvert\, s_x$ | $a$ が負でも標準偏差は正(絶対値) |

**例**:ある都市の気温(摂氏)の平均が 20℃、標準偏差が 5℃ のとき、華氏 $F = 1.8C + 32$ に直すと

- 平均:$1.8 \times 20 + 32 = 68$°F
- 標準偏差:$1.8 \times 5 = 9$°F(+32 は散らばりに影響しない)

全員の点数に 10 点を足しても、散らばり(分散・標準偏差)は変わらない、と考えるとイメージしやすくなります。

## 標準化(z 得点)と偏差値

平均を引いて標準偏差で割る変換を **標準化** といい、変換後の値を **z 得点**(標準得点)と呼びます。

$$z_i = \frac{x_i - \bar{x}}{s}$$

- 標準化したデータは、**平均 0・標準偏差 1** になる(上の変換の性質で $a = 1/s$, $b = -\bar{x}/s$ とした場合)
- z 得点は「平均から標準偏差何個分離れているか」を表すので、**単位や平均点が違う値どうしを比較** できる

**偏差値** は、z 得点を平均 50・標準偏差 10 になるように変換したものです。

$$\text{偏差値} = 50 + 10 z$$

**例**:平均 60 点、標準偏差 10 点のテストで 75 点なら、$z = (75 - 60) \div 10 = 1.5$、偏差値は $50 + 10 \times 1.5 = 65$ です。

分布が正規分布に近いとき、z 得点(偏差値)と「その範囲に入る人の割合」は次の図のように対応します。

![正規分布と z 得点・偏差値の対応](images/u05_hensachi.png)

- 平均 ± 1 標準偏差(偏差値 40〜60)に約 68%、± 2 標準偏差(偏差値 30〜70)に約 95% が入る
- 偏差値 60 以上は上位約 16%、偏差値 70 以上は上位約 2.3%

## 変動係数

平均の大きさが大きく違うデータどうしでは、標準偏差をそのまま比べても散らばりの大小を正しく比べられません。
そこで、標準偏差を平均で割った **変動係数**(CV)を使います。

$$CV = \frac{s}{\bar{x}}$$

**例**:ゾウとネズミの体重

| | 平均 | 標準偏差 | 変動係数 |
|---|---|---|---|
| ゾウ | 5000 kg | 400 kg | 400 ÷ 5000 = 0.08 |
| ネズミ | 0.02 kg | 0.004 kg | 0.004 ÷ 0.02 = 0.20 |

標準偏差はゾウの方がはるかに大きいですが、平均に対する相対的な散らばりは、ネズミの方が大きいことがわかります。

- 単位がない(単位が打ち消し合う)ので、**単位の違うデータどうしも比較できる**
- 平均が正の値をとる、比例尺度のデータ(身長・体重・金額など)に使う。平均が 0 に近いデータや、負の値をとるデータには向かない

## 指数化

時点や地域によって大きさが違う系列を比べるとき、**基準となる時点の値を 100** として各時点の値を表すことを **指数化** といいます。

$$\text{指数}_t = \frac{x_t}{x_{\text{基準}}} \times 100$$

**例**:ある商品の価格(円)を 2021 年 = 100 として指数化する

| 年 | 2021 | 2022 | 2023 | 2024 | 2025 |
|---|---|---|---|---|---|
| 価格 | 180 | 186 | 195 | 210 | 231 |
| 指数 | 100.0 | 103.3 | 108.3 | 116.7 | 128.3 |

- 指数 128.3 は「基準時点より 28.3% 高い」ことを表す
- 2024 年から 2025 年への変化率は、**指数の比** から $128.3 \div 116.7 - 1 \approx 0.10$(10% 上昇)。元の価格の比 $231 \div 210 - 1 = 0.10$ と同じ
- **指数の差(128.3 − 116.7 = 11.6 ポイント)は変化率ではない** 点に注意(基準年からの変化を「ポイント」で表したもの)

消費者物価指数(CPI)や株価指数も、指数化の考え方を使っています。

## まとめ

- $y = ax + b$ と変換すると、平均は $a\bar{x} + b$、標準偏差は $\lvert a \rvert$ 倍。足し算は散らばりに影響しない。
- z 得点は平均 0・標準偏差 1。偏差値は $50 + 10z$。単位や平均が違う値の比較に使う。正規分布なら偏差値 60 以上は上位約 16%。
- 変動係数 $s/\bar{x}$ は単位のない散らばりの指標。平均の大きさが違うデータの比較に使う。
- 指数化は基準時点を 100 とする表し方。変化率は指数の比から求め、指数の差(ポイント)と混同しない。
