# 単回帰と予測
1つの変数からもう1つの変数を予測する回帰直線を、最小二乗法で求めます。回帰係数・決定係数・分散分析表を手計算で求め、残差プロットや統計ソフトの出力を読み取れるようにします。

## 学習目標
- 最小二乗法による回帰係数・切片を、平均・分散・共分散から手計算し、予測に使える
- 変動の分解(全変動 = 回帰変動 + 残差変動)と決定係数を説明・計算できる
- 回帰の分散分析表を作り、F 値と残差の標準誤差を求められる
- 残差プロットと統計ソフトの出力を読み取り、変数変換の考え方を説明できる

## 回帰直線と最小二乗法

説明変数 $x$ から目的変数(被説明変数)$y$ を予測する直線 $\hat{y} = a + b x$ を **回帰直線**、$a$ を切片、$b$ を **回帰係数**(傾き)といいます。
実際の値と予測値の差 $e_i = y_i - \hat{y}_i$ を **残差** といい、**残差の2乗和を最小にする** ように $a, b$ を決める方法を **最小二乗法** といいます。

$$b = \frac{s_{xy}}{s_x^2} = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sum (x_i - \bar{x})^2}, \qquad a = \bar{y} - b\,\bar{x}$$

**例**:「散布図と相関」の単元で相関係数を求めた 5 組のデータ($\bar{x} = 3$, $\bar{y} = 4$)

- 偏差の積の和 $\sum (x - \bar{x})(y - \bar{y}) = 6$、$x$ の偏差平方和 $\sum (x - \bar{x})^2 = 10$
- 回帰係数 $b = 6 \div 10 = 0.6$、切片 $a = 4 - 0.6 \times 3 = 2.2$
- 回帰直線は $\hat{y} = 2.2 + 0.6x$

| $x$ | 1 | 2 | 3 | 4 | 5 | 合計 |
|---|---|---|---|---|---|---|
| $y$ | 2 | 4 | 5 | 4 | 5 | |
| 予測値 $\hat{y}$ | 2.8 | 3.4 | 4.0 | 4.6 | 5.2 | |
| 残差 $y - \hat{y}$ | −0.8 | 0.6 | 1.0 | −0.6 | −0.2 | 0 |

![回帰直線と残差](images/u08_fit_residuals.png)

回帰直線の性質:
- 必ず平均の点 $(\bar{x}, \bar{y})$ を通る(図の × 印)
- 残差の合計は 0 になる
- $b = r \dfrac{s_y}{s_x}$(相関係数と標準偏差で表せる)。$b$ と $r$ の符号は一致する
- $x$ で $y$ を予測する直線と、$y$ で $x$ を予測する直線は一般に **一致しない**

**予測の注意**:データの範囲外の $x$ で予測する **外挿** は、直線関係が成り立つ保証がありません。また、回帰係数 $b$ は「$x$ が 1 増えると、$y$ は **平均的に** $b$ 増える」という関係を表すもので、因果関係を示すものではありません。

## 変動の分解と決定係数

$y$ のばらつき(全変動)は、回帰直線で説明できる部分と説明できない部分(残差)に分けられます。

$$\underbrace{\sum (y_i - \bar{y})^2}_{\text{全変動 } S_T} = \underbrace{\sum (\hat{y}_i - \bar{y})^2}_{\text{回帰変動 } S_R} + \underbrace{\sum (y_i - \hat{y}_i)^2}_{\text{残差変動 } S_E}$$

全変動のうち回帰で説明できる割合が **決定係数** です。

$$R^2 = \frac{S_R}{S_T} = 1 - \frac{S_E}{S_T}$$

上の例では、$S_T = 4 + 0 + 1 + 0 + 1 = 6$、$S_E = 0.64 + 0.36 + 1 + 0.36 + 0.04 = 2.4$、$S_R = 6 - 2.4 = 3.6$ なので、$R^2 = 3.6 \div 6 = 0.6$ です。

- $0 \le R^2 \le 1$。1 に近いほど当てはまりがよい
- **単回帰では $R^2 = r^2$**(相関係数の2乗)。上の例では $r \approx 0.775$ で、$0.775^2 \approx 0.6$。逆に $r = -0.8$ なら $R^2 = 0.64$

## 回帰の分散分析表と残差の標準誤差

変動の分解を、自由度とともに表にしたものが **分散分析表** です($n$:データ数)。上の例($n = 5$)で作ると次のようになります。

| 要因 | 平方和 | 自由度 | 平均平方 | F 値 |
|---|---|---|---|---|
| 回帰 | $S_R = 3.6$ | 1 | $V_R = 3.6 \div 1 = 3.6$ | $F = V_R \div V_E = 4.5$ |
| 残差 | $S_E = 2.4$ | $n - 2 = 3$ | $V_E = 2.4 \div 3 = 0.8$ | |
| 全体 | $S_T = 6$ | $n - 1 = 4$ | | |

- 残差の自由度が $n - 2$ なのは、$a$ と $b$ の2つを推定しているため
- $\sqrt{V_E}$ を **残差の標準誤差**(回帰の標準誤差)といい、予測値のまわりの典型的なずれの大きさを表す(単位は $y$ と同じ)。上の例では $\sqrt{0.8} \approx 0.89$
- F 値が大きいほど、回帰で説明できる部分が残差に比べて大きい(F 値を使った検定は「回帰と実験計画」の講座で扱います)

## 残差プロット

横軸に予測値、縦軸に残差をとった **残差プロット** で、回帰直線の当てはまりを確認します。

![残差プロットの典型的なパターン](images/u08_residual_patterns.png)

| パターン | 意味 | 対応 |
|---|---|---|
| 0 のまわりに一様にばらつく(左) | 当てはまりはよい | そのまま使える |
| 曲線的なパターン(中央) | 関係が直線でない | 変数変換や2次の項を検討 |
| 予測値が大きいほどばらつきが広がる(右) | ばらつきが一定でない | 対数変換などを検討 |

## 統計ソフトの出力を読む

試験では、統計ソフトの回帰分析の出力から値を読み取る問題がよく出ます。次は、生徒 120 人の数学の点数(math)を週あたりの学習時間(study_hours)で回帰した出力の一部です。

```
                 coef    std err        t
const         35.7489      1.948   18.348
study_hours    2.1528      0.241    8.933

R-squared:  0.403    Adj. R-squared:  0.398
F-statistic: 79.80   No. Observations:  120
```

| 出力の項目 | 意味 | この出力では |
|---|---|---|
| `coef`(`const` の行) | 切片 $a$ | 35.75 |
| `coef`(説明変数の行) | 回帰係数 $b$ | 学習時間が週 1 時間多いと、数学は平均的に約 2.15 点高い |
| `std err` | 係数の標準誤差(推定のばらつき) | |
| `t` | 係数 ÷ 標準誤差(係数が 0 かどうかの検定に使う。「回帰と実験計画」の講座で扱う) | 2.1528 ÷ 0.241 ≈ 8.93 |
| `R-squared` | 決定係数 | 数学の点数の変動の約 40% を学習時間で説明できる |
| `Adj. R-squared` | 自由度調整済み決定係数 | |
| `F-statistic` | 分散分析表の F 値 | |
| `No. Observations` | データ数 $n$ | 120 |

この回帰直線を使うと、学習時間が週 10 時間の生徒の数学の予測値は $35.75 + 2.153 \times 10 \approx 57.3$ 点です。

## 変数変換

関係が直線でないときは、変数を変換すると直線で表せることがあります。

| 関係 | 変換 | 直線になる式 |
|---|---|---|
| 一定の **率** で増える(指数的) $y = c\, e^{bx}$ | $y$ を対数に | $\log y = \log c + b x$ |
| べき乗の関係 $y = c\, x^b$ | 両方を対数に | $\log y = \log c + b \log x$ |

$\log y = a + bt$($\log$ は自然対数)のとき、$t$ が 1 増えると $y$ は $e^b$ 倍になります。つまり、1期あたりの成長率は $e^b - 1$ です。

次の図は、あるサービスの会員数(12 か月分)を、そのまま回帰した場合(左)と、対数をとって回帰した場合(右)の比較です。

![対数変換で指数的な増加が直線になる](images/u08_log_transform.png)

右の図では点が直線にほぼ完全に乗り、決定係数も大きくなります。傾きは $b \approx 0.0867$ なので、1 か月あたりの成長率は $e^{0.0867} - 1 \approx 0.091$(約 9.1%)です。

ある商品の売上高 $y$ を時点 $t$ に対して、自然対数をとって回帰したところ、$\log y = 4.2 + 0.05\,t$ となった。1 期あたりの売上高の成長率(%)を求めよ(小数第2位まで)。

## まとめ

- 最小二乗法:$b = s_{xy}/s_x^2$(偏差の積の和 ÷ $x$ の偏差平方和)、$a = \bar{y} - b\bar{x}$。回帰直線は $(\bar{x}, \bar{y})$ を通り、残差の合計は 0。
- 全変動 = 回帰変動 + 残差変動。決定係数 $R^2 = S_R/S_T$ は単回帰では $r^2$ に等しい。
- 分散分析表:回帰の自由度 1、残差の自由度 $n-2$。$F = V_R / V_E$、残差の標準誤差は $\sqrt{V_E}$。
- 残差プロットでパターン(曲線・広がり)がないかを確認する。一定の率で増える関係は対数変換で直線になり、成長率は $e^b - 1$。
- 統計ソフトの出力では、const の coef が切片、説明変数の coef が回帰係数、R-squared が決定係数。外挿や因果の解釈には注意する。
