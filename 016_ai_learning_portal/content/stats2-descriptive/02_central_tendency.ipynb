# 中心傾向の指標:平均値・中央値・最頻値
分布の「中心」を1つの値で表す3つの代表値を学びます。分布の形と代表値の大小関係を読み取り、度数分布表だけから代表値を手計算で求められるようにします。

## 学習目標
- 平均値・中央値・最頻値の定義と性質を説明し、少ないデータから手計算で求められる
- 分布の歪みと3つの代表値の大小関係を、グラフから読み取れる
- 度数分布表(階級値と度数)だけから平均値・中央値の階級・最頻値を求められる
- グループごとの平均から、全体の平均(加重平均)を正しく求められる

## 3つの代表値

| 代表値 | 定義 | 特徴 |
|---|---|---|
| **平均値** $\bar{x}$ | $\bar{x} = \dfrac{1}{n}\sum_{i=1}^{n} x_i$(合計 ÷ 個数) | すべての値を使う。外れ値の影響を強く受ける |
| **中央値**(メディアン) | 小さい順に並べたときの真ん中の値。$n$ が偶数なら真ん中の2つの平均 | 外れ値の影響を受けにくい(頑健) |
| **最頻値**(モード) | 最も多く現れる値(度数分布表では、度数最大の階級の階級値) | 質的変数にも使える。連続データでは階級の取り方で変わる |

**例**:7 個のデータ `3, 5, 5, 6, 8, 9, 40` の代表値を求めます。

| 代表値 | 求め方 | 値 |
|---|---|---|
| 平均値 | 合計 76 ÷ 7 | 約 10.86 |
| 中央値 | 小さい順に並べて 4 番目 | 6 |
| 最頻値 | 2 回現れる値 | 5 |

最後の 40 は他の値から大きく離れた **外れ値** です。40 を除いた 6 個で求め直すと、平均値は 6.0、中央値は 5.5 になります。
外れ値1つで平均値は 4.86 も動きましたが、中央値は 0.5 しか動いていません。これが「中央値は外れ値に強い」ということです。

**平均値の性質**:各データの平均からの差(**偏差** $x_i - \bar{x}$)を合計すると、必ず 0 になります。
上の例の偏差は −7.86, −5.86, −5.86, −4.86, −2.86, −1.86, +29.14 で、合計は 0 です。平均値は、データの「重心」(つり合いの位置)だと考えることができます。

## 分布の形と代表値の大小関係

分布の形によって、3つの代表値の並び方が決まります。試験で頻出のポイントです。

![分布の形と最頻値・中央値・平均値の位置](images/u02_shapes_mmm.png)

| 分布の形 | 大小関係 | 理由 |
|---|---|---|
| 右に裾が長い | 最頻値 < 中央値 < 平均値 | 右側の大きな値が平均値を引き上げる |
| 左右対称(単峰) | 最頻値 ≒ 中央値 ≒ 平均値 | 中心が一致する |
| 左に裾が長い | 平均値 < 中央値 < 最頻値 | 左側の小さな値が平均値を引き下げる |

覚え方:**平均値は裾の側に引っぱられる**。最頻値は山の頂上、中央値はその間にあります。

実際のデータでも確かめてみます。次の図は、200 世帯の年収のヒストグラムに3つの代表値を重ねたものです。

![年収のヒストグラムと平均値・中央値・最頻値](images/u02_income_lines.png)

右に裾が長い分布なので、最頻値(450 万円)< 中央値(482 万円)< 平均値(544.5 万円)の順に並んでいます。
年収の「平均」が多くの人の実感より高くなるのはこのためで、**典型的な値を示すには中央値が適しています**。

## 度数分布表から代表値を求める

元のデータがなく度数分布表だけが与えられたときは、各階級のデータがすべて **階級値** にあるとみなして近似します。

**例**:ある店の来店客 40 人の滞在時間(分)

| 階級(分) | 階級値 $m$ | 度数 $f$ | $m \times f$ | 累積度数 |
|---|---|---|---|---|
| 0 以上 10 未満 | 5 | 6 | 30 | 6 |
| 10 以上 20 未満 | 15 | 15 | 225 | 21 |
| 20 以上 30 未満 | 25 | 9 | 225 | 30 |
| 30 以上 40 未満 | 35 | 7 | 245 | 37 |
| 40 以上 50 未満 | 45 | 3 | 135 | 40 |
| 計 | | 40 | 860 | |

- **平均値**:$\bar{x} \approx \dfrac{\sum m f}{n} = \dfrac{860}{40} = 21.5$ 分
- **中央値**:40 人の真ん中は 20 番目と 21 番目。累積度数が 6 → 21 なので、どちらも **10〜20 分の階級** に含まれる。
- **最頻値**:度数が最大(15)の階級は 10〜20 分なので、その階級値 **15 分**。

より詳しく(参考):中央値の階級の中でデータが均等に並んでいると考えると、中央値は
$10 + \dfrac{20 - 6}{15} \times 10 \approx 19.3$ 分 と見積もれます(階級の下限 + 中央までの残りの人数 ÷ 階級の度数 × 階級幅)。

## 加重平均:グループの平均から全体の平均を求める

グループごとの平均値と人数がわかっているとき、全体の平均は **人数で重みをつけた平均(加重平均)** になります。

$$\bar{x} = \frac{n_1\bar{x}_1 + n_2\bar{x}_2 + \cdots + n_k\bar{x}_k}{n_1 + n_2 + \cdots + n_k}$$

**例**:A 店(客 30 人、平均購入額 2,000 円)と B 店(客 10 人、平均購入額 4,000 円)を合わせた平均購入額は

$$\frac{30 \times 2000 + 10 \times 4000}{30 + 10} = \frac{100000}{40} = 2500 \text{ 円}$$

2 店の平均を単純に平均した 3,000 円(**平均の平均**)とは一致しません。人数の多い A 店の重みが大きいためです。グループの大きさが違うときに平均の平均を使うのは誤りです。

## 外れ値に強い平均:トリム平均

上下の極端な値を一定割合だけ取り除いてから平均する **トリム平均**(刈り込み平均)は、平均値と中央値の中間的な性質を持ちます。
採点競技で最高点と最低点を除いて平均するのも同じ考え方です。上の年収データ(200 世帯)では、上下 10% ずつ(20 世帯ずつ)を除いたトリム平均は約 498.3 万円で、平均値(544.5 万円)と中央値(482 万円)の間になります。

## まとめ

- 平均値は外れ値に弱く、中央値は頑健。最頻値は質的変数にも使える。偏差の合計は必ず 0。
- 平均値は裾の側に引っぱられる:右に裾が長い分布では「最頻値 < 中央値 < 平均値」、左に裾が長いと逆。
- 度数分布表だけのときは、Σ(階級値 × 度数)÷ 総数で平均を近似し、累積度数で中央値の階級を探す。最頻値は度数最大の階級の階級値。
- グループの平均を統合するときは、人数で重みをつけた加重平均を使う。
