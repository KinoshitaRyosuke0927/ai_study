# データの種類と分布の表し方
データの種類(質的変数・量的変数)を見分け、度数分布表・ヒストグラム・幹葉図などから分布の姿を読み取ります。統計検定2級の「データソース」「データの分布の記述」に対応します。

## 学習目標
- 変数を質的変数(名義・順序)と量的変数(離散・連続)に分類できる
- 度数分布表(階級値・相対度数・累積相対度数)の意味を説明し、表から割合や中央値の階級を読み取れる
- ヒストグラム・累積度数グラフ・棒グラフ・幹葉図の特徴を理解し、目的に合わせて使い分けられる
- 分布の形状(右に裾が長い・対称・単峰・多峰など)をグラフから読み取り、言葉で説明できる

## データソース:どこからデータを得るか

分析は「どのデータを使うか」から始まります。自分で集めたデータ(一次データ)だけでなく、他の機関が集めて公開しているデータ(二次データ)も重要です。

| 種類 | 例 |
|---|---|
| 公的統計(国・自治体が作成) | 国勢調査、家計調査、労働力調査、消費者物価指数 |
| 公的統計の窓口 | 政府統計の総合窓口 **e-Stat** |
| 業務データ | 売上・アクセスログ・アンケート |

二次データを使うときは、**調査の対象・時点・定義**(例:「世帯」の定義、税込か税抜か)を必ず確認します。定義が違うデータどうしを比べると、誤った結論になります。

この講座では、次の4つのデータを使います(いずれも学習用に作成した架空のデータです)。

| ファイル | 内容 |
|---|---|
| `households.csv` | 200 世帯の地域・世帯人数・年収(万円)・住居・生活満足度(1〜5) |
| `students.csv` | 120 人の学習時間(週・時間)・数学・英語・理科の点数・睡眠時間 |
| `kids.csv` | 小学 1〜6 年生 90 人の足の大きさ(cm)・語彙数 |
| `monthly_sales.csv` | 2018 年 1 月〜2025 年 12 月の月次売上(百万円) |

## 変数の種類

変数は、値の性質によって次のように分類します。どの集計・グラフ・代表値が使えるかは、変数の種類で決まります。

| 分類 | 細分類 | 特徴 | 例 | 使える代表値 |
|---|---|---|---|---|
| **質的変数**(カテゴリカル・データ) | 名義尺度 | 区別するだけ。順序がない | 血液型、都道府県、職業 | 最頻値 |
| | 順序尺度 | 順序はあるが、間隔に意味がない | 成績の段階(A/B/C)、アンケートの5段階評価 | 最頻値・中央値 |
| **量的変数** | 離散型 | とびとびの値(数えられる) | 来店者数、子どもの人数 | 平均・中央値・最頻値 |
| | 連続型 | 連続的な値(測る) | 身長、気温、所要時間 | 平均・中央値 |

**見分け方**:次の順に問いかけると、迷わず分類できます。

1. 値の **足し算・平均に意味があるか**? → ない:質的変数 / ある:量的変数
2. (質的変数なら)カテゴリに **順序があるか**? → ある:順序尺度 / ない:名義尺度
3. (量的変数なら)値を **数えるか、測るか**? → 数える:離散型 / 測る:連続型

注意点:
- **数字で記録されていても質的変数のことがあります**(5段階評価、郵便番号、会員番号)。数字かどうかではなく、平均に意味があるかで判断します。
- 量的変数は、さらに **間隔尺度**(0 に絶対的な意味がない。温度 ℃ など)と **比例尺度**(0 が「無い」を表し、比が意味を持つ。年収・身長など)に分けることもあります。「20℃は10℃の2倍暑い」とは言えないのが間隔尺度です。

## 度数分布表

量的変数は、値をいくつかの **階級** に区切り、各階級に入るデータの個数(**度数**)を数えて **度数分布表** にまとめます。次の表は、`students.csv` の 120 人の数学の点数をまとめたものです。

| 階級(点) | 階級値 | 度数 | 相対度数 | 累積度数 | 累積相対度数 |
|---|---|---|---|---|---|
| 10 以上 20 未満 | 15 | 1 | 0.008 | 1 | 0.008 |
| 20 以上 30 未満 | 25 | 3 | 0.025 | 4 | 0.033 |
| 30 以上 40 未満 | 35 | 18 | 0.150 | 22 | 0.183 |
| 40 以上 50 未満 | 45 | 30 | 0.250 | 52 | 0.433 |
| 50 以上 60 未満 | 55 | 36 | 0.300 | 88 | 0.733 |
| 60 以上 70 未満 | 65 | 23 | 0.192 | 111 | 0.925 |
| 70 以上 80 未満 | 75 | 7 | 0.058 | 118 | 0.983 |
| 80 以上 90 未満 | 85 | 2 | 0.017 | 120 | 1.000 |
| 計 | | 120 | 1.000 | | |

| 用語 | 意味 | 表での例 |
|---|---|---|
| 階級・階級幅 | 値を区切った区間とその幅 | 「50 以上 60 未満」、幅 10 点 |
| 階級値 | 階級の中央の値(階級を代表する値) | 50〜60 点の階級値は 55 点 |
| 度数 | 階級に入るデータの個数 | 50〜60 点は 36 人 |
| 相対度数 | 度数 ÷ データの総数 | 36 ÷ 120 = 0.300 |
| 累積度数・累積相対度数 | 最初の階級からその階級までの度数・相対度数の合計 | 60 点未満は 88 人、0.733 |

**表の読み方**(試験でよく問われる読み取り):

- 「60 点未満の生徒の割合」は、50〜60 点の階級の **累積相対度数** を読めばよい → 0.733(約 73%)。
- 「70 点以上の生徒の割合」は、1 − (60〜70 点の累積相対度数) = 1 − 0.925 = 0.075(7.5%)。
- **中央値** は、累積相対度数が初めて 0.5 以上になる階級に含まれる → 50〜60 点の階級(0.433 → 0.733)。
- 度数が最も大きい階級は 50〜60 点で、その階級値 55 点を **最頻値** とみなす。

階級の作り方の注意:
- 階級の数の目安として **スタージェスの公式** $k \approx 1 + \log_2 n$ があります($n = 120$ なら約 7.9 で、上の表の 8 階級に近い)。
- 境界の値をどちらの階級に入れるか(「以上・未満」か「より大きい・以下」か)を必ず決めて表に書きます。

## ヒストグラムと累積度数グラフ

度数分布表を柱のグラフにしたものが **ヒストグラム**、累積相対度数を折れ線にしたものが **累積度数グラフ** です。下の図は、上の度数分布表を図にしたものです。

![数学の点数のヒストグラムと累積相対度数グラフ](images/u01_hist_cum.png)

- **ヒストグラム**:分布の山の位置・広がり・形がひと目でわかる。ここでは 50〜60 点を頂上とする、ほぼ左右対称の山です。
- **累積度数グラフ**:縦軸の値が「その点数未満の割合」を表す。縦軸 0.5 の位置から横に線を引き、グラフと交わる点の横軸を読むと、**中央値のおよその値**(約 52 点)がわかります。同じように、縦軸 0.25・0.75 の位置から四分位数のおよその値も読めます。

**ヒストグラムは柱の「面積」が度数を表す** ことに注意します。階級幅がそろっていれば高さ=度数ですが、幅が違う階級を混ぜるときは高さを調整します。例えば上の表で 70〜90 点を1つの階級(幅 20 点、度数 7 + 2 = 9)にまとめるなら、幅が2倍なので、柱の高さは 9 ÷ 2 = 4.5(幅 10 点あたりの度数)で描きます。

**階級幅を変えると見え方が変わる**:次の図は、`households.csv` の 200 世帯の年収を、階級幅だけを変えて描いたものです。

![同じ年収データを階級幅 25・100・800 で描いたヒストグラム](images/u01_binwidth.png)

- 幅が **狭すぎる**(左)と、度数が少ない階級が多くなり、でこぼこが目立って全体の形がつかみにくい。
- 幅が **広すぎる**(右)と、山や裾がつぶれ、「右に裾が長い」という特徴が見えなくなる。
- 中央くらいの幅にすると、500 万円前後を山とし、高い年収の側に長く裾を引く形がよくわかります。

ヒストグラムを読むとき・描くときは、階級幅が適切かを意識しましょう。

## 質的変数のグラフ:棒グラフと円グラフ

![地域別の世帯数の棒グラフと円グラフ](images/u01_bar_pie.png)

| グラフ | 向いていること | 向いていないこと |
|---|---|---|
| **棒グラフ** | カテゴリどうしの大小の比較(関東が他の約2倍、など) | — |
| **円グラフ** | 全体に占める割合(構成比)を示す | カテゴリが多いとき、複数の円グラフどうしの比較 |

**ヒストグラムと棒グラフの違い**:見た目は似ていますが、ヒストグラムは **量的変数** の分布を表すため、柱の間をあけず、横軸の並び順と幅に意味があります。棒グラフは質的変数のカテゴリごとの度数を表し、カテゴリの並び順は自由(名義尺度の場合)です。

## 幹葉図

**幹葉図**(stem-and-leaf plot)は、数値の上の桁を「幹」、下の桁を「葉」として並べた図です。ヒストグラムのように分布の形がわかり、しかも **元の値をすべて読み取れる** のが特徴です。

例:8 個のデータ `52, 55, 61, 63, 63, 68, 70, 74` を、十の位を幹、一の位を葉として表すと次のようになります。

```
5 | 2 5
6 | 1 3 3 8
7 | 0 4
```

**読み方**:
- 各行の葉の数がその階級(50 点台・60 点台・70 点台)の度数 → 横向きのヒストグラムとして形が見える。
- 元の値が残っているので、代表値を正確に求められる。小さい順に数えて、データの個数 8 の真ん中(4 番目と 5 番目)は 63 と 63 なので **中央値は 63**、最も多く現れる **最頻値も 63**、**範囲** は 74 − 52 = 22。
- データの個数が奇数なら真ん中の1つ、偶数なら真ん中の2つの平均が中央値です。

## 分布の形状

ヒストグラムの形は、次のような言葉で表します。試験では、グラフと説明文の対応がよく問われます。

![分布の形の例(左から 右に裾が長い・左に裾が長い・左右対称・一様・二峰)](images/u01_shape_examples.png)

| 形状 | 特徴 | 例 |
|---|---|---|
| 右に裾が長い(右に歪んだ) | 山が左寄りで、大きい値の側に長く裾を引く | 年収、貯蓄額、待ち時間 |
| 左に裾が長い(左に歪んだ) | 山が右寄りで、小さい値の側に長く裾を引く | 易しいテストの点数、寿命 |
| 左右対称・ベル型 | 中心から左右対称の山 | 身長、測定誤差 |
| 一様 | どの階級もほぼ同じ度数 | サイコロの目 |
| 単峰・多峰 | 山が1つ/2つ以上 | 多峰:性別や年代の異なる集団が混ざったデータ |

「右に裾が長い」は **裾が伸びている向き** で呼ぶ点に注意します(山の位置は左寄り)。
多峰の分布は、**性質の違う集団が混ざっている** サインです。層(グループ)に分けて見直すと、それぞれは単峰になることがよくあります。

度数分布とグラフについて述べた次の記述のうち、**誤っているもの** を1つ選べ。

- ヒストグラムでは、各階級の度数は柱の面積で表される。
- 階級幅が他の階級の 2 倍の階級があるとき、その階級の柱の高さは、度数の 1/2 にあたる高さで描く。
- 地域別の世帯数のような質的変数の度数を比べるには、ヒストグラムではなく棒グラフを用いる。
- 累積相対度数グラフで縦軸が 0.5 となる横軸の値から、中央値のおよその値を読み取れる。
- 階級幅を変えても、ヒストグラムから読み取れる分布の形は変わらない。

## まとめ

- 変数の種類(名義・順序・離散・連続)によって、使える集計方法と代表値が決まる。
- 度数分布表は、階級の境界の扱い(以上・未満)と階級幅を明示する。累積相対度数から「〜以下の割合」や中央値の位置がわかる。
- ヒストグラムは量的変数、棒グラフ・円グラフは質的変数に使う。幹葉図は元の値を残したまま分布の形を示せる。
- 分布の形状(歪み・峰の数)を言葉で説明できるようにしておく。多峰性は異なる集団の混在を疑う。
