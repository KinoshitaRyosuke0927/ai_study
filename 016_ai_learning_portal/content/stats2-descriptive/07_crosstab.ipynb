# カテゴリカルデータとクロス集計
質的変数どうしの関係を、度数表と2元クロス表で調べます。どの合計を分母にした割合を見るべきかを意識して、表やグラフを正しく読み取れるようにします。

## 学習目標
- 質的変数の度数表・相対度数表を読み取れる
- 2元クロス表(分割表)の周辺度数を理解し、表から割合を手計算できる
- 行パーセント・列パーセント・全体パーセントを、答えたい問いに合わせて使い分けられる
- 帯グラフでグループ間の構成比を比べ、独立なときの期待度数を求められる

## 度数表

質的変数は、カテゴリごとの度数と相対度数を表にまとめます。

- 名義尺度:度数の多い順に並べると読みやすい
- 順序尺度:**カテゴリの順序どおり** に並べる。累積相対度数にも意味がある

**例**:200 世帯の生活満足度(1:不満 〜 5:満足)の度数表

| 満足度 | 度数 | 相対度数 | 累積相対度数 |
|---|---|---|---|
| 1 | 5 | 0.025 | 0.025 |
| 2 | 44 | 0.220 | 0.245 |
| 3 | 78 | 0.390 | 0.635 |
| 4 | 54 | 0.270 | 0.905 |
| 5 | 19 | 0.095 | 1.000 |

順序があるので、「満足度 2 以下の世帯は 24.5%」のように累積相対度数で読めます。中央値(100・101 番目)は満足度 3 です。

## 2元クロス表

2つの質的変数の組み合わせごとに度数を数えた表を **2元クロス表**(分割表)といいます。行と列の合計を **周辺度数** といいます。

**例**:200 世帯の地域と住居のクロス表

| 地域 | 持ち家 | 賃貸 | 計 |
|---|---|---|---|
| 北日本 | 17 | 13 | 30 |
| 関東 | 32 | 35 | 67 |
| 中部 | 19 | 13 | 32 |
| 近畿 | 24 | 13 | 37 |
| 西日本 | 19 | 15 | 34 |
| 計 | 111 | 89 | 200 |

最下行の「計」(111, 89)と右端の列の「計」(30, 67, …)が周辺度数、右下の 200 が総数です。

## 何を分母にするか:行パーセントと列パーセント

クロス表を割合にするときは、**何を分母にするか** で意味が変わります。上の表の「関東の賃貸 35 世帯」で考えます。

| 割合 | 分母 | 計算 | 答えられる問い |
|---|---|---|---|
| 行パーセント | 行の合計(関東 67) | 35 ÷ 67 = 52.2% | 関東の世帯のうち、賃貸は何% か |
| 列パーセント | 列の合計(賃貸 89) | 35 ÷ 89 = 39.3% | 賃貸の世帯のうち、関東は何% か |
| 全体パーセント | 総数(200) | 35 ÷ 200 = 17.5% | 全世帯のうち、関東の賃貸は何% か |

問題文の「**〜のうち**」が分母を表しています。
地域によって持ち家率が違うかを知りたいときは、**比べたいグループ(地域)を分母にする行パーセント** を比べます。

| 地域 | 持ち家率(行パーセント) |
|---|---|
| 北日本 | 56.7% |
| 関東 | 47.8% |
| 中部 | 59.4% |
| 近畿 | 64.9% |
| 西日本 | 55.9% |
| 全体 | 55.5% |

関東だけが、全体の持ち家率(55.5%)を大きく下回っていることがわかります。

## 帯グラフ

グループごとの構成比を比べるには、各グループを長さ 100% の帯にして積み上げた **帯グラフ**(100% 積み上げ棒グラフ)が便利です。円グラフを並べるより比較しやすくなります。

![地域別の住居の構成比(帯グラフ)](images/u07_band.png)

帯グラフは **割合** を表すので、帯の長さからはグループの人数(度数)の違いはわかりません。人数を比べたいときは、各グループの人数を添えるか、度数の棒グラフを使います。

## 独立なときの期待度数

2つの変数に関係がない(**独立** である)とき、各マスの度数は周辺度数から次のように予想できます。

$$E_{ij} = \frac{(\text{行 } i \text{ の合計}) \times (\text{列 } j \text{ の合計})}{\text{総数}}$$

例えば、地域と住居が独立なら、関東の持ち家の期待度数は $67 \times 111 \div 200 \approx 37.2$ 世帯です。実際は 32 世帯なので、関東は「独立なら予想される数より持ち家が少ない」ことになります。

観測度数と期待度数のずれが偶然の範囲を超えているかは、「仮説検定」の講座で扱う **独立性の検定**(カイ二乗検定)で調べます。

## まとめ

- 順序尺度の度数表はカテゴリの順に並べる。累積相対度数で「〜以下の割合」が読める。
- 2元クロス表の行・列の合計を周辺度数という。
- 割合の分母(行・列・全体)によって答えられる問いが変わる。「〜のうち」が分母。比べたいグループを分母にする。
- 帯グラフは構成比の比較に向くが、人数(度数)はわからない。独立なときの期待度数は「行の合計 × 列の合計 ÷ 総数」。
