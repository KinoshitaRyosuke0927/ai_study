# 散らばりの指標と箱ひげ図
データが中心のまわりにどれだけ散らばっているかを、範囲・分散・標準偏差・四分位範囲で表します。手計算で求める手順と、箱ひげ図の読み方を身につけます。

## 学習目標
- 分散(n − 1 で割る不偏分散)と標準偏差を、定義式と「2乗の平均 − 平均の2乗」の両方で手計算できる
- n で割る分散と n − 1 で割る分散の違いを説明できる
- 四分位数・四分位範囲・5数要約を求め、1.5×IQR の基準で外れ値を判定できる
- 箱ひげ図を読み取り、ヒストグラムとの対応や複数の分布の違いを説明できる

## 範囲

**範囲**(レンジ)= 最大値 − 最小値。計算は簡単ですが、両端の2つの値だけで決まるため、外れ値が1つあるだけで大きく変わります。

## 分散と標準偏差

各データの平均からの差 $x_i - \bar{x}$ を **偏差** といいます。偏差の合計は必ず 0 になるので、2乗してから合計します(**偏差平方和**)。

$$S = \sum_{i=1}^{n}(x_i - \bar{x})^2$$

| 指標 | 定義 | 備考 |
|---|---|---|
| 分散(n で割る) | $\dfrac{S}{n}$ | データそのものの散らばり(標本分散と呼ぶこともある) |
| **不偏分散**(n − 1 で割る) | $s^2 = \dfrac{S}{n-1}$ | 母分散の推定に使う。**2級の出題範囲表では「分散(n−1で割る)」** |
| 標準偏差 | 分散の平方根 | 元のデータと同じ単位になる |

**例**:5 個のデータ `4, 7, 7, 9, 13`(平均 8)の分散を手計算で求めます。

| $x$ | 4 | 7 | 7 | 9 | 13 | 合計 |
|---|---|---|---|---|---|---|
| 偏差 $x - \bar{x}$ | −4 | −1 | −1 | 1 | 5 | 0 |
| 偏差の2乗 | 16 | 1 | 1 | 1 | 25 | **44** |

- n で割る分散:44 ÷ 5 = **8.8**
- 不偏分散:44 ÷ 4 = **11**、標準偏差は $\sqrt{11} \approx 3.32$

**計算を楽にする公式**:n で割る分散は「2乗の平均 − 平均の2乗」でも求められます。

$$\frac{1}{n}\sum x_i^2 - \bar{x}^2$$

上の例では、2乗の平均が $(16 + 49 + 49 + 81 + 169) \div 5 = 72.8$、平均の2乗が $8^2 = 64$ なので、72.8 − 64 = 8.8 となり一致します。
合計 $\sum x$ と2乗の合計 $\sum x^2$ だけが与えられる問題では、この公式を使います。

なぜ n − 1 で割るのか:偏差は「標本の平均 $\bar{x}$」からの差なので、母平均からの差より小さめになります。n − 1 で割るとこの偏りが補正されます(詳しくは「推定」の講座で扱います)。n − 1 を **自由度** といいます。

## 四分位数と四分位範囲

データを小さい順に並べて4等分する位置の値を **四分位数** といいます。

- 第1四分位数 $Q_1$(下位 25%)、第2四分位数 $Q_2$(= 中央値)、第3四分位数 $Q_3$(下位 75%)
- **四分位範囲** IQR $= Q_3 - Q_1$:中央の 50% のデータが入る幅。外れ値の影響を受けにくい
- 四分位偏差 $= \mathrm{IQR} \div 2$

**求め方**(この講座の約束):中央値でデータを下半分と上半分に分け、それぞれの中央値を $Q_1$・$Q_3$ とします(データの個数が奇数のときは、中央値そのものはどちらにも含めません)。
四分位数の求め方にはほかの流儀もあるので、試験では問題文の定義に従います。

**例**:10 個のデータ `2, 4, 5, 7, 8, 10, 12, 13, 15, 30`

| 手順 | 内容 | 値 |
|---|---|---|
| 中央値 | 5 番目と 6 番目(8 と 10)の平均 | 9 |
| $Q_1$ | 下半分 `2, 4, 5, 7, 8` の中央値 | 5 |
| $Q_3$ | 上半分 `10, 12, 13, 15, 30` の中央値 | 13 |
| IQR | 13 − 5 | 8 |

**外れ値の判定**:$Q_1 - 1.5\,\mathrm{IQR}$ より小さい値、または $Q_3 + 1.5\,\mathrm{IQR}$ より大きい値を外れ値とみなします。
この例では、下側の基準は 5 − 12 = −7、上側の基準は 13 + 12 = 25 なので、**30 が外れ値** です。

## 5数要約と箱ひげ図

**5数要約**(最小値・$Q_1$・中央値・$Q_3$・最大値)を図にしたものが **箱ひげ図** です。上の例を箱ひげ図にすると次のようになります。

![箱ひげ図の各部の名前](images/u03_boxplot_parts.png)

- 箱:$Q_1$ から $Q_3$ まで(幅が IQR)。箱の中の線が中央値
- ひげ:箱の端から、**外れ値の基準の内側にある最も遠いデータ** まで(この例では、上側のひげは 30 ではなく 15 まで)
- 外れ値:基準の外側の値を点で描く

ひげを単純に最小値・最大値まで伸ばす描き方もあります。また、**箱ひげ図からは平均値は読み取れない** 点に注意しましょう(平均を ◆ などで追加して描くことはあります)。

## ヒストグラムと箱ひげ図の対応

同じデータのヒストグラム(上)と箱ひげ図(下)を並べると、分布の形が箱ひげ図のどこに現れるかがわかります。

![ヒストグラムと箱ひげ図の対応](images/u03_box_hist.png)

| 分布の形 | 箱ひげ図の特徴 |
|---|---|
| 右に裾が長い | 右側のひげが長く、中央値が箱の左寄り。右側に外れ値が出やすい |
| 左右対称 | 左右のひげの長さがほぼ同じで、中央値が箱のほぼ中央 |
| 左に裾が長い | 左側のひげが長く、中央値が箱の右寄り。左側に外れ値が出やすい |

## 2つの分布を比べる

平均が同じでも、散らばりが違えば分布の姿はまったく異なります。分布を比べるときは、**中心(平均・中央値)と散らばり(標準偏差・IQR)の両方** を見ます。

![平均が同じで散らばりが違う2つの分布](images/u03_spread_compare.png)

A と B の平均はどちらも約 60 点ですが、標準偏差は A が約 4.6 点、B が約 14.6 点です。B の方が、高い点・低い点の生徒がともに多くいます。

比べるときのコツ:
- 複数のヒストグラムは **横軸の目盛りをそろえて** 縦に並べる(目盛りが違うと見た目で誤解する)
- グループが多いときは、箱ひげ図を並べると中心・散らばり・外れ値を一度に比べられる

## まとめ

- 分散は偏差平方和を n − 1(不偏分散)または n で割ったもの。n で割る分散は「2乗の平均 − 平均の2乗」でも求められる。
- 標準偏差は分散の平方根で、元のデータと同じ単位。外れ値の影響を受けやすい。
- 四分位範囲は中央 50% の幅で、外れ値に強い。外れ値の目安は $Q_1 - 1.5\,\mathrm{IQR}$ 未満・$Q_3 + 1.5\,\mathrm{IQR}$ 超。
- 箱ひげ図は5数要約の図。裾が長い側にひげが伸びる。平均値は読み取れない。分布の比較では中心と散らばりの両方を見る。
