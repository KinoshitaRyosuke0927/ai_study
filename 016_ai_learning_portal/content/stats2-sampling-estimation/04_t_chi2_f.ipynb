# t 分布・カイ二乗分布・F 分布
正規母集団からの標本で、推定や検定に使う3つの分布の成り立ちと性質を学びます。分布表から上側確率点を読み取り、どの統計量がどの分布に従うかを使い分けられるようにします。

## 学習目標
- カイ二乗分布・t 分布・F 分布が、正規分布からどのように作られるかを説明できる
- カイ二乗分布の平均・分散、t 分布と標準正規分布の関係を説明できる
- 分布表から上側確率点を読み取り、F 分布の下側確率点を求められる
- 標本平均・不偏分散・分散の比がどの分布に従うかを対応づけられる

## 3つの分布の形

## カイ二乗分布

$Z_1, \dots, Z_k$ が独立に標準正規分布に従うとき、その2乗の和

$$\chi^2 = Z_1^2 + Z_2^2 + \cdots + Z_k^2$$

は **自由度 $k$ のカイ二乗分布** $\chi^2(k)$ に従います。

- 0 以上の値をとり、右に裾が長い
- **平均は $k$、分散は $2k$**
- 正規母集団からの標本では、$\dfrac{(n-1)s^2}{\sigma^2} \sim \chi^2(n - 1)$(母分散の推定・検定に使う)

## t 分布

$Z \sim N(0, 1)$ と $W \sim \chi^2(k)$ が独立なとき

$$T = \frac{Z}{\sqrt{W / k}}$$

は **自由度 $k$ の t 分布** $t(k)$ に従います。

- 0 を中心に左右対称で、標準正規分布より裾が重い
- 自由度が大きくなると標準正規分布に近づく
- 正規母集団からの標本では、$\dfrac{\bar{X} - \mu}{s/\sqrt{n}} \sim t(n - 1)$(母分散が未知のときの母平均の推定・検定に使う)

## F 分布

$U \sim \chi^2(m)$ と $V \sim \chi^2(n)$ が独立なとき

$$F = \frac{U / m}{V / n}$$

は **自由度 $(m, n)$ の F 分布** $F(m, n)$ に従います。

- 0 以上の値をとり、右に裾が長い
- 2つの正規母集団の分散が等しいとき、$s_1^2 / s_2^2 \sim F(n_1 - 1, n_2 - 1)$(分散の比の推定・検定に使う)
- 下側確率点は、自由度を入れ替えた上側確率点の逆数:$F_{1-\alpha}(m, n) = \dfrac{1}{F_\alpha(n, m)}$
- $T \sim t(k)$ なら $T^2 \sim F(1, k)$(例:$t_{0.025}(10)^2 = 2.228^2 \approx 4.965 = F_{0.05}(1, 10)$)

## 分布表の読み方

試験では、必要な上側確率点が表で与えられます。行が自由度、列が上側確率です。

| 自由度 | $t$:上側 5% | $t$:上側 2.5% | $\chi^2$:下側 2.5% | $\chi^2$:上側 5% | $\chi^2$:上側 2.5% |
|---|---|---|---|---|---|
| 5 | 2.015 | 2.571 | 0.831 | 11.070 | 12.833 |
| 10 | 1.812 | 2.228 | 3.247 | 18.307 | 20.483 |
| 15 | — | — | 6.262 | 24.996 | 27.488 |
| 30 | 1.697 | 2.042 | — | — | — |
| ∞(標準正規) | 1.645 | 1.960 | — | — | — |

F 分布の上側 5% 点の例:$F_{0.05}(5, 10) = 3.326$、$F_{0.05}(10, 5) = 4.735$。
自由度 (5, 10) の F 分布の **下側** 5% 点は、$1 / F_{0.05}(10, 5) = 1 / 4.735 \approx 0.211$ です。

**例**:正規母集団($\sigma^2 = 4$)から大きさ 11 の標本をとったとき、不偏分散 $s^2$ が大きくなりすぎる確率を考えます。$10 s^2 / 4 \sim \chi^2(10)$ なので、$P(10 s^2 / 4 > 18.307) = 0.05$、つまり $s^2 > 7.32$ となる確率は 5% です。

t 分布・カイ二乗分布・F 分布について述べた次の記述のうち、**誤っているもの** を1つ選べ。

- t 分布は 0 を中心に左右対称で、自由度が大きくなると標準正規分布に近づく。
- カイ二乗分布は、負の値をとることもある左右対称な分布である。
- 自由度 k のカイ二乗分布の平均は k である。
- 自由度 (m, n) の F 分布の下側確率点は、自由度 (n, m) の F 分布の上側確率点の逆数で求められる。
- 自由度 k の t 分布に従う確率変数の2乗は、自由度 (1, k) の F 分布に従う。

## まとめ

- カイ二乗分布 $\chi^2(k)$:標準正規分布の2乗の和。平均 $k$、分散 $2k$。$(n-1)s^2/\sigma^2 \sim \chi^2(n-1)$。
- t 分布 $t(k)$:$Z/\sqrt{W/k}$。左右対称で裾が重く、自由度が大きいと標準正規分布に近づく。$(\bar{X} - \mu)/(s/\sqrt{n}) \sim t(n-1)$。
- F 分布 $F(m, n)$:カイ二乗分布の比。$s_1^2/s_2^2 \sim F(n_1 - 1, n_2 - 1)$。下側確率点は自由度を入れ替えた上側確率点の逆数。$t(k)^2 = F(1, k)$。
