<!-- @question

正規母集団からの大きさ $n$ の標本について、**自由度 $n - 1$ の t 分布** に従う統計量を選べ($s$ は不偏分散の平方根)。

- (X̄ − μ)/(σ/√n)
- (X̄ − μ)/(s/√n)
- (n − 1)s²/σ²
- s²/σ²
