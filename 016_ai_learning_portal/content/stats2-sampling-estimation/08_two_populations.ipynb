# 2つの母集団の区間推定
2つの母集団の母平均の差・母分散の比・母比率の差を区間で推定します。信頼区間と仮説検定の関係(区間が 0 や 1 を含むかどうか)もあわせて身につけます。

## 学習目標
- 母平均の差の信頼区間を、母分散が既知の場合と等分散を仮定する場合で求められる
- 母分散の比の信頼区間を、F 分布のパーセント点を使って求められる
- 母比率の差の信頼区間を求められる
- 信頼区間から、両側検定の結論を読み取れる

## 母平均の差の区間推定

**母分散が既知** の場合:

$$(\bar{x}_1 - \bar{x}_2) \pm z_{\alpha/2}\sqrt{\frac{\sigma_1^2}{n_1} + \frac{\sigma_2^2}{n_2}}$$

**母分散が未知で、等しいと仮定できる** 場合は、プールした分散 $s_p^2 = \dfrac{(n_1-1)s_1^2 + (n_2-1)s_2^2}{n_1+n_2-2}$ と、自由度 $n_1 + n_2 - 2$ の t 分布を使います。

$$(\bar{x}_1 - \bar{x}_2) \pm t_{\alpha/2}(n_1 + n_2 - 2)\sqrt{s_p^2\left(\frac{1}{n_1} + \frac{1}{n_2}\right)}$$

**例**(「仮説検定」の講座の教え方の比較と同じデータ):新しい教え方のクラス(10 人、平均 75、不偏分散 36)と従来のクラス(12 人、平均 70、不偏分散 40)。

- $s_p^2 = (9 \times 36 + 11 \times 40) / 20 = 38.2$、標準誤差 $\sqrt{38.2 \times (1/10 + 1/12)} \approx 2.646$
- 自由度 20 の上側 2.5% 点は 2.086
- $5 \pm 2.086 \times 2.646$ → **[−0.52, 10.52]**

## 信頼区間と検定の関係

**母平均の差の 95% 信頼区間が 0 を含む** ことと、**「差がない」という帰無仮説が両側 5% の検定で棄却されない** ことは同じです。

![信頼区間が 0 を含むかどうかと検定の結論の対応](images/u08_ci_test.png)

上の教え方の例は区間が 0 を含むので、両側 5% の t 検定で棄却されません(「仮説検定」の講座の結論と一致します)。
信頼区間は、検定の結論に加えて **差がどのくらいの大きさか** もわかるので、検定より多くの情報を持っています。

## 母分散の比の区間推定

$\dfrac{s_1^2/\sigma_1^2}{s_2^2/\sigma_2^2} \sim F(n_1 - 1, n_2 - 1)$ を使います。母分散の比 $\sigma_1^2/\sigma_2^2$ の 95% 信頼区間は

$$\frac{s_1^2/s_2^2}{F_{0.025}(n_1 - 1, n_2 - 1)} \leq \frac{\sigma_1^2}{\sigma_2^2} \leq \frac{s_1^2}{s_2^2} \times F_{0.025}(n_2 - 1, n_1 - 1)$$

上限の F 分布は、**自由度を入れ替えたもの** です(下側 2.5% 点の逆数が、自由度を入れ替えた上側 2.5% 点になるため)。

**例**:$n_1 = 11$, $s_1^2 = 24$、$n_2 = 16$, $s_2^2 = 10$ のとき、$s_1^2/s_2^2 = 2.4$。$F_{0.025}(10, 15) = 3.060$、$F_{0.025}(15, 10) = 3.522$。

- 下限 $2.4 / 3.060 \approx 0.78$、上限 $2.4 \times 3.522 \approx 8.45$ → **[0.78, 8.45]**
- 区間が **1 を含む** ので、「分散が等しい」という帰無仮説は両側 5% の F 検定で棄却されない

## 母比率の差の区間推定

$$(\hat{p}_1 - \hat{p}_2) \pm z_{\alpha/2}\sqrt{\frac{\hat{p}_1(1 - \hat{p}_1)}{n_1} + \frac{\hat{p}_2(1 - \hat{p}_2)}{n_2}}$$

検定(帰無仮説のもとで比率が等しいと考え、プールした比率を使う)と違い、区間推定では **それぞれの標本比率で標準誤差を計算** します。

**例**:広告 A の購入率 60/200 = 0.30、広告 B の購入率 45/200 = 0.225。

- 標準誤差 $\sqrt{0.3 \times 0.7 / 200 + 0.225 \times 0.775 / 200} \approx 0.0438$
- $0.075 \pm 1.96 \times 0.0438$ → **[−0.011, 0.161]**(0 を含む)

## まとめ

- 母平均の差:σ 既知なら $z$、等分散なら $s_p^2$ と自由度 $n_1 + n_2 - 2$ の $t$ を使う。
- 母分散の比:下限は $F_{0.025}(n_1-1, n_2-1)$ で割り、上限は自由度を入れ替えた $F_{0.025}(n_2-1, n_1-1)$ を掛ける。
- 母比率の差:区間推定ではそれぞれの標本比率で標準誤差を計算する(検定ではプールした比率)。
- 差の区間が 0 を含む(比の区間が 1 を含む)⇔ 両側検定で棄却されない。区間は差の大きさも示す。
