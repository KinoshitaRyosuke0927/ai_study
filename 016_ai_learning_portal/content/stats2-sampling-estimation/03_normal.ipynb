# 正規分布と標準正規分布表
推測統計で最もよく使う正規分布について、標準化と標準正規分布表を使って確率を求める方法を身につけます。正規分布の和の性質(再生性)と、二項分布の正規近似・連続修正も扱います。

## 学習目標
- 正規分布に従う確率変数を標準化し、標準正規分布表から確率を求められる
- 上側確率点(パーセント点)の意味を説明し、正規分布の上側確率点を計算できる
- 独立な正規分布の和・差の分布(再生性)を使って確率を求められる
- 二項分布の確率を、連続修正を使って正規近似できる

## 正規分布と標準化

平均 $\mu$、分散 $\sigma^2$ の正規分布を $N(\mu, \sigma^2)$ と書きます。左右対称のベル型で、次の性質があります。

- 平均 ± 1σ に約 68.3%、± 2σ に約 95.4%、± 3σ に約 99.7% が入る
- $X \sim N(\mu, \sigma^2)$ を **標準化** した $Z = \dfrac{X - \mu}{\sigma}$ は、**標準正規分布** $N(0, 1)$ に従う

標準化すれば、どんな正規分布の確率も、標準正規分布の表1つで求められます。

標準正規分布表(上側確率 $P(Z > z)$)の抜粋:

| $z$ | 0.5 | 1.0 | 1.28 | 1.5 | 1.645 | 1.96 | 2.0 | 2.326 | 2.5 | 2.576 | 3.0 |
|---|---|---|---|---|---|---|---|---|---|---|---|
| $P(Z > z)$ | 0.3085 | 0.1587 | 0.1003 | 0.0668 | 0.0500 | 0.0250 | 0.0228 | 0.0100 | 0.0062 | 0.0050 | 0.0013 |

**例**:$X \sim N(60, 10^2)$ のとき

![X ~ N(60, 10²) と標準化した Z の対応](images/u03_standardize.png)

| 求めたい確率 | 標準化 | 値 |
|---|---|---|
| $P(X > 75)$ | $P(Z > (75 - 60)/10) = P(Z > 1.5)$ | 0.0668 |
| $P(X < 50)$ | $P(Z < -1) = P(Z > 1)$(対称性) | 0.1587 |
| $P(50 < X < 70)$ | $P(-1 < Z < 1) = 1 - 2 \times 0.1587$ | 0.6826 |

表は上側確率だけなので、**左右対称** であることと **全体の確率が 1** であることを使って、ほかの確率に読み替えます。

## 上側確率点

確率変数が $c$ を超える確率が $\alpha$ になるとき、$c$ を **上側 $100\alpha\%$ 点**(上側確率点・パーセント点)といいます。標準正規分布の上側 $\alpha$ 点を $z_\alpha$ と書きます。

| $\alpha$ | 0.05 | 0.025 | 0.01 | 0.005 |
|---|---|---|---|---|
| $z_\alpha$ | 1.645 | 1.96 | 2.326 | 2.576 |

$X \sim N(\mu, \sigma^2)$ の上側 $\alpha$ 点は、$\mu + z_\alpha \sigma$ です。例えば $N(60, 10^2)$ の上側 5% 点は $60 + 1.645 \times 10 = 76.45$ です。

## 正規分布の再生性

$X \sim N(\mu_1, \sigma_1^2)$ と $Y \sim N(\mu_2, \sigma_2^2)$ が独立なとき、和と差も正規分布に従います。

$$X + Y \sim N(\mu_1 + \mu_2,\ \sigma_1^2 + \sigma_2^2), \qquad X - Y \sim N(\mu_1 - \mu_2,\ \sigma_1^2 + \sigma_2^2)$$

**差でも分散は足し算** になる点に注意します。

**例**:$X \sim N(50, 3^2)$ と $Y \sim N(40, 4^2)$ が独立なら、$X + Y \sim N(90, 25)$(標準偏差 5)、$X - Y \sim N(10, 25)$(標準偏差 5)です。

## 二項分布の正規近似と連続修正

二項分布 $B(n, p)$(成功確率 $p$ の試行を $n$ 回行ったときの成功回数)は、平均 $np$、分散 $np(1-p)$ です。$n$ が大きいとき、正規分布 $N(np, np(1-p))$ で近似できます。

二項分布は整数の値しかとらないので、正規分布で近似するときは **境界を 0.5 ずらす**(**連続修正**)と精度がよくなります。

**例**:公平なコインを 100 回投げて、表が 60 回以上出る確率

- 平均 $100 \times 0.5 = 50$、標準偏差 $\sqrt{100 \times 0.5 \times 0.5} = 5$
- 「60 回以上」は、整数 60 の柱の左端 59.5 から右の面積と考える

![二項分布 B(100, 0.5) と正規近似・連続修正](images/u03_binom_approx.png)

$$P(X \geq 60) \approx P\left(Z \geq \frac{59.5 - 50}{5}\right) = P(Z \geq 1.9) \approx 0.0287$$

二項分布で正確に計算すると 0.0284 で、よく一致します(連続修正をしないと $P(Z \geq 2) = 0.0228$ となり、ずれが大きくなります)。

| 求めたい確率 | 連続修正後 |
|---|---|
| $P(X \geq 60)$ | $P(X \geq 59.5)$ |
| $P(X \leq 40)$ | $P(X \leq 40.5)$ |
| $P(X > 60) = P(X \geq 61)$ | $P(X \geq 60.5)$ |

## まとめ

- $X \sim N(\mu, \sigma^2)$ は $Z = (X - \mu)/\sigma$ で標準化し、標準正規分布表で確率を求める。対称性と「全体で 1」を使って読み替える。
- 上側 α 点:$z_{0.05} = 1.645$、$z_{0.025} = 1.96$、$z_{0.01} = 2.326$、$z_{0.005} = 2.576$。$N(\mu, \sigma^2)$ の上側 α 点は $\mu + z_\alpha\sigma$。
- 独立な正規分布の和・差は正規分布で、平均は和・差、分散はどちらも和。
- 二項分布 $B(n, p)$ は $N(np, np(1-p))$ で近似できる。境界を 0.5 ずらす連続修正で精度が上がる。
