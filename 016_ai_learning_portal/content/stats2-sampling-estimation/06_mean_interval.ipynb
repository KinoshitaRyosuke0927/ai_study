# 母平均の区間推定
母平均を「幅をもった区間」で推定する区間推定を学びます。母分散が既知・未知の場合の信頼区間の手計算、信頼係数の正しい意味、必要な標本の大きさの求め方を身につけます。

## 学習目標
- 母分散が既知のとき、母平均の信頼区間を標準正規分布で求められる
- 母分散が未知のとき、t 分布を使って母平均の信頼区間を求められる
- 信頼係数 95% の意味を正しく説明できる
- 信頼区間の幅と標本の大きさ・信頼係数の関係を説明し、必要な標本の大きさを計算できる

## 母分散が既知の場合

母標準偏差 $\sigma$ が既知の正規母集団からの標本では、$\dfrac{\bar{X} - \mu}{\sigma/\sqrt{n}} \sim N(0, 1)$ なので

$$P\left(-1.96 \leq \frac{\bar{X} - \mu}{\sigma/\sqrt{n}} \leq 1.96\right) = 0.95$$

これを $\mu$ について解くと、母平均の **95% 信頼区間** が得られます。

$$\bar{x} - 1.96\frac{\sigma}{\sqrt{n}} \leq \mu \leq \bar{x} + 1.96\frac{\sigma}{\sqrt{n}}$$

一般に、信頼係数 $1 - \alpha$ の信頼区間は $\bar{x} \pm z_{\alpha/2}\,\sigma/\sqrt{n}$ です(99% なら $z_{0.005} = 2.576$)。

**例**:成人男性 36 人の身長の平均は 172.3 cm、母標準偏差は 6 cm とわかっている。

- 標準誤差 $6 / \sqrt{36} = 1$
- 95% 信頼区間:$172.3 \pm 1.96 \times 1$ → **[170.34, 174.26]**
- 99% 信頼区間:$172.3 \pm 2.576 \times 1$ → [169.72, 174.88](信頼係数を上げると区間は広くなる)

## 信頼係数の意味

「95% 信頼区間」とは、**同じ方法で標本をとって区間を作ることをくり返すと、そのうち約 95% の区間が本当の母平均を含む** という意味です。

次の図は、母平均 50 の母集団から 20 回標本をとり、それぞれ 95% 信頼区間を作ったものです。

![同じ母集団から作った 20 個の 95% 信頼区間](images/u06_ci_sim.png)

20 個のうち 19 個が母平均 50 を含み、1 個(赤)は含んでいません。

- 母平均 $\mu$ は決まった値(未知の定数)で、変わるのは標本ごとの区間の方
- したがって、「計算した区間 [170.34, 174.26] に μ が 95% の確率で含まれる」という言い方は厳密には正しくない(1つの区間は μ を含むか含まないかのどちらか)

## 母分散が未知の場合

母標準偏差がわからないときは、不偏分散の平方根 $s$ で代用し、自由度 $n - 1$ の t 分布を使います。

$$\bar{x} \pm t_{\alpha/2}(n - 1)\,\frac{s}{\sqrt{n}}$$

**例**:ある部品 16 個の長さの平均は 50.2 mm、標準偏差 $s = 4$ mm だった。母平均の 95% 信頼区間を求める。

- 標準誤差 $4 / \sqrt{16} = 1$、自由度 15 の上側 2.5% 点は 2.131
- $50.2 \pm 2.131 \times 1$ → **[48.07, 52.33]**

t 分布のパーセント点は標準正規分布のもの(1.96)より大きいので、σ が既知の場合より区間は広くなります。

## 区間の幅と標本の大きさ

95% 信頼区間の幅(上限 − 下限)は $2 \times 1.96 \times \sigma/\sqrt{n}$ です。

| 変えること | 区間の幅 |
|---|---|
| 標本の大きさ $n$ を大きくする | 狭くなる($n$ を 4 倍にすると半分) |
| 信頼係数を上げる(95% → 99%) | 広くなる |
| 母標準偏差 $\sigma$ が大きい | 広くなる |

**必要な標本の大きさ**:推定の誤差(区間の半分の幅)を $E$ 以下にしたいとき

$$1.96\frac{\sigma}{\sqrt{n}} \leq E \quad\Longrightarrow\quad n \geq \left(\frac{1.96\,\sigma}{E}\right)^2$$

**例**:$\sigma = 10$ のとき、95% 信頼区間の半分の幅を 2 以下にするには、$n \geq (1.96 \times 10 / 2)^2 = 96.04$ なので、**97 以上** にします(切り上げる)。

## まとめ

- σ 既知:$\bar{x} \pm z_{\alpha/2}\,\sigma/\sqrt{n}$(95% なら 1.96、99% なら 2.576)。
- σ 未知:$\bar{x} \pm t_{\alpha/2}(n-1)\,s/\sqrt{n}$。t のパーセント点は z より大きく、区間は広くなる。
- 95% 信頼区間は「くり返し作ると約 95% が母平均を含む」区間。母平均は定数で、変わるのは区間。
- 幅は n を 4 倍で半分、信頼係数を上げると広がる。必要な標本の大きさは $n \geq (z_{\alpha/2}\sigma/E)^2$(切り上げ)。
