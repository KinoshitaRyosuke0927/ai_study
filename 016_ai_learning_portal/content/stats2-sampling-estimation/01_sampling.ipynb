# 母集団と標本・標本平均の分布
推測統計の出発点となる、母集団と標本、母数と統計量の関係を学びます。標本平均がどのような分布をするか(期待値・分散・標準誤差)を理解し、手計算で求められるようにします。

## 学習目標
- 母集団・標本・母数・統計量の違いを説明し、記号(μ と x̄ など)を正しく使い分けられる
- 標本平均の期待値が μ、分散が σ²/n であることを説明し、標準誤差を計算できる
- 標本の大きさと標準誤差の関係(n を 4 倍にすると標準誤差は半分)を使える
- 有限母集団から非復元抽出するときの修正を説明できる

## 母集団と標本

| 用語 | 意味 | 例 |
|---|---|---|
| **母集団** | 調べたい対象全体 | 日本の成人男性全員 |
| **標本** | 母集団から取り出した一部 | 無作為に選んだ 100 人 |
| **母数** | 母集団の特徴を表す値(ふつう未知) | 母平均 $\mu$、母分散 $\sigma^2$、母比率 $p$ |
| **統計量** | 標本から計算する値(標本ごとに変わる) | 標本平均 $\bar{x}$、不偏分散 $s^2$、標本比率 $\hat{p}$ |

推測統計では、**統計量(手元にある値)から母数(知りたい値)を推し量ります**。そのために、標本を **無作為に**(どの対象も同じ確率で選ばれるように)取り出すことが前提になります。

統計量は、どの標本が選ばれるかによって値が変わる **確率変数** です。その確率分布を **標本分布** といいます。

## 標本平均の期待値と分散

母平均 $\mu$、母分散 $\sigma^2$ の母集団から、大きさ $n$ の標本 $X_1, \dots, X_n$ を独立に取り出すとき、標本平均 $\bar{X} = \dfrac{1}{n}\sum X_i$ は次の性質を持ちます。

$$E[\bar{X}] = \mu, \qquad V[\bar{X}] = \frac{\sigma^2}{n}$$

- 期待値:標本平均は、平均すると母平均に一致する(かたよりがない)
- 分散:独立な $X_i$ の和の分散は $n\sigma^2$。それを $n$ で割るので、分散は $n\sigma^2 / n^2 = \sigma^2/n$

標本平均の標準偏差 $\sigma/\sqrt{n}$ を **標準誤差** といいます。

**例**:身長の母平均が 170 cm、母標準偏差が 6 cm の母集団から 9 人を選ぶとき、標本平均の期待値は 170 cm、標準誤差は $6 / \sqrt{9} = 2$ cm です。

次の図は、右に裾が長い母集団(母平均 10、母標準偏差 5)から標本をとったときの、標本平均の分布です。

![母集団の分布と、標本平均の分布(n = 4・16)](images/u01_sampling_means.png)

- どの $n$ でも、標本平均の分布の中心は母平均 10
- $n$ が大きいほど分布は細くなる:標準誤差は $n = 4$ で 2.5、$n = 16$ で 1.25
- **標準誤差を半分にするには、$n$ を 4 倍** にする必要がある($\sqrt{n}$ に反比例するため)

## 標本比率の分布

成功の確率が $p$ の試行を $n$ 回行ったとき、成功の割合(標本比率)$\hat{p}$ の期待値と分散は

$$E[\hat{p}] = p, \qquad V[\hat{p}] = \frac{p(1-p)}{n}$$

です。例えば $p = 0.3$、$n = 100$ なら、標本比率の標準偏差は $\sqrt{0.3 \times 0.7 / 100} \approx 0.046$ です。

## 有限母集団からの非復元抽出

上の公式は、母集団が十分大きい(または取り出した対象を戻す **復元抽出**)ときのものです。
大きさ $N$ の有限母集団から、戻さずに $n$ 個取り出す **非復元抽出** では、分散に **有限母集団修正** がかかります。

$$V[\bar{X}] = \frac{N - n}{N - 1} \cdot \frac{\sigma^2}{n}$$

**例**:$N = 100$ 人のクラスから $n = 20$ 人を選ぶ($\sigma^2 = 25$)とき、修正係数は $80 / 99 \approx 0.808$ で、分散は $0.808 \times 25 / 20 \approx 1.01$ です(修正しなければ 1.25)。

- $N$ が $n$ に比べて十分大きければ、修正係数はほぼ 1 で無視できる
- 全数調査($n = N$)なら修正係数は 0 で、標本平均は母平均に一致する(ばらつかない)

## まとめ

- 母数(μ, σ², p)は母集団の値でふつう未知。統計量(x̄, s², p̂)は標本から計算し、標本ごとに変わる確率変数。
- 標本平均は $E[\bar{X}] = \mu$、$V[\bar{X}] = \sigma^2/n$。標準誤差は $\sigma/\sqrt{n}$ で、半分にするには n を 4 倍にする。
- 標本比率は $E[\hat{p}] = p$、$V[\hat{p}] = p(1-p)/n$。
- 有限母集団から非復元抽出するときは、分散に $(N - n)/(N - 1)$ を掛ける。
