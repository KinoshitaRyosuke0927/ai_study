# 点推定と推定量の性質
母数を1つの値で推定する点推定と、よい推定量の条件(不偏性・一致性・有効性)を学びます。推定量の期待値と分散を計算して、推定量を比べられるようにします。

## 学習目標
- 推定量と推定値の違いを説明できる
- 推定量の期待値を計算し、不偏推定量かどうかを判定できる
- n で割る分散がかたよる理由と、不偏分散が n − 1 で割る理由を説明できる
- 一致性・有効性の意味を説明し、推定量の分散を比べられる

## 推定量と推定値

母数 $\theta$(μ や σ² など)を、標本から計算した1つの値で推定することを **点推定** といいます。

| 用語 | 意味 | 例 |
|---|---|---|
| **推定量** | 推定に使う計算方法(標本 $X_1, \dots, X_n$ の関数)。確率変数 | $\bar{X} = \dfrac{1}{n}\sum X_i$ |
| **推定値** | 実際に得られた標本で計算した具体的な数値 | $\bar{x} = 172.3$ |

推定量は確率変数なので、期待値や分散を考えることができます。これを使って「よい推定量」の条件を考えます。

![推定量のかたよりとばらつきのイメージ](images/u05_bias_variance.png)

図の中心(+)が本当の母数、点が標本ごとの推定値です。推定値の中心が本当の値からずれていることを **かたより**(バイアス)、推定値の散らばりを **ばらつき**(分散)といいます。

## 不偏性

推定量の期待値が母数に一致するとき、その推定量を **不偏推定量** といいます。

$$E[\hat{\theta}] = \theta$$

| 母数 | 不偏推定量 | 理由 |
|---|---|---|
| 母平均 $\mu$ | 標本平均 $\bar{X}$ | $E[\bar{X}] = \mu$ |
| 母比率 $p$ | 標本比率 $\hat{p}$ | $E[\hat{p}] = p$ |
| 母分散 $\sigma^2$ | 不偏分散 $s^2 = \dfrac{1}{n-1}\sum (X_i - \bar{X})^2$ | $E[s^2] = \sigma^2$ |

**n で割る分散はかたよる**:$\dfrac{1}{n}\sum (X_i - \bar{X})^2$ の期待値は $\dfrac{n-1}{n}\sigma^2$ で、母分散より小さくなります。偏差を「標本平均 $\bar{X}$」から測っているため、母平均から測るより小さめになるからです。

次の図は、母分散 4 の正規母集団から $n = 5$ の標本を何度もとり、2通りの方法で分散を推定した結果です。

![n − 1 で割る分散と n で割る分散の分布](images/u05_unbiased_var.png)

n − 1 で割ると推定値の平均は真の値 4 に一致しますが、n で割ると $\frac{4}{5} \times 4 = 3.2$ 付近にかたよります。

**推定量の期待値の計算例**:$X_1, X_2$ が母平均 $\mu$ の母集団からの標本のとき、$T = \dfrac{X_1 + 2X_2}{3}$ の期待値は $\dfrac{\mu + 2\mu}{3} = \mu$ なので、$T$ も不偏推定量です。

## 一致性

標本の大きさ $n$ を大きくすると、推定量が母数に近づく性質を **一致性** といいます(そのような推定量を **一致推定量** といいます)。

- 標本平均 $\bar{X}$ は、大数の法則により μ の一致推定量
- n で割る分散は、かたよりはあるが、$\frac{n-1}{n} \to 1$ なので σ² の一致推定量

不偏性と一致性は別の性質です。不偏でも一致しない推定量(例:常に $X_1$ だけを使う)もあります。

## 有効性

不偏推定量どうしなら、**分散が小さい方がよい**(推定値が本当の値の近くに集まる)と考えます。これを **有効性** といい、分散が小さい方を「より有効な推定量」といいます。

**例**:母分散 $\sigma^2 = 9$ の母集団からの標本 $X_1, X_2$ で、2つの不偏推定量を比べる。

| 推定量 | 期待値 | 分散 |
|---|---|---|
| $\dfrac{X_1 + X_2}{2}$ | $\mu$ | $\dfrac{1}{4}(9 + 9) = 4.5$ |
| $\dfrac{X_1 + 2X_2}{3}$ | $\mu$ | $\dfrac{1}{9}(9 + 4 \times 9) = 5$ |

独立な確率変数の和の分散は、係数の **2乗** を掛けて足します($V[aX_1 + bX_2] = a^2 V[X_1] + b^2 V[X_2]$)。どちらも不偏ですが、分散が小さい $\dfrac{X_1 + X_2}{2}$(標本平均)の方が有効です。

## まとめ

- 推定量は計算方法(確率変数)、推定値は実際に得た数値。
- 不偏性:$E[\hat{\theta}] = \theta$。$\bar{X}$・$\hat{p}$・不偏分散 $s^2$ は不偏推定量。n で割る分散の期待値は $\frac{n-1}{n}\sigma^2$ でかたよる。
- 一致性:n を大きくすると母数に近づく。有効性:不偏推定量のうち分散が小さいほどよい。
- 推定量の分散は $V[aX_1 + bX_2] = a^2 V[X_1] + b^2 V[X_2]$(独立なとき)で計算する。
