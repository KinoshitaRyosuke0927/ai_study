# 分野別練習1:記述統計
「統計検定2級対策 1:記述統計」の範囲から、本番と同じ5択形式の問題を8問出題します。1問ずつ採点され、ヒントと解説を確認できます。必要な数値は問題文に示します。

<!-- @question
type: choice
answer: A と Y、B と Z、C と X
title: ヒストグラムと箱ひげ図の対応
hint: 右に裾が長い分布では、箱ひげ図の右のひげが長く、中央値は箱の左寄りになります。
explanation: A は右に裾が長く、右のひげが長い Y と対応します。B は左右対称で、箱とひげがほぼ対称な Z と対応します。C は左に裾が長く、左のひげが長く外れ値も左側にある X と対応します。
-->
次の A〜C は3つのデータのヒストグラム、X〜Z は同じ3つのデータの箱ひげ図を、順不同で並べたものである。

![ヒストグラム A〜C と箱ひげ図 X〜Z](images/u1_hist_box.png)

ヒストグラムと箱ひげ図の対応として正しいものを選べ。

- A と X、B と Y、C と Z
- A と X、B と Z、C と Y
- A と Y、B と X、C と Z
- A と Y、B と Z、C と X
- A と Z、B と Y、C と X

<!-- @question
type: choice
answer: 最頻値 < 中央値 < 平均値
title: 代表値の大小関係
hint: 右に裾が長い分布では、一部の大きな値に平均値が引っ張られます。
explanation: 右に裾が長い分布では、少数の非常に大きな値に引っ張られて平均値が最も大きくなり、山の頂上にあたる最頻値が最も小さくなります。中央値はその間です。
-->
ある国の世帯の年間所得の分布は、右に裾が長い分布である。この分布の平均値・中央値・最頻値の大小関係として、最も適切なものを選べ。

- 平均値 < 中央値 < 最頻値
- 最頻値 < 中央値 < 平均値
- 中央値 < 平均値 < 最頻値
- 平均値 = 中央値 = 最頻値
- 中央値 < 最頻値 < 平均値

製品 A の重さは平均 50 g・標準偏差 8 g、製品 B の重さは平均 120 g・標準偏差 15 g であった。ばらつきの比較として最も適切なものを選べ。

- 標準偏差が大きいので、B の方が相対的なばらつきが大きい。
- 変動係数は A が 0.16、B が 0.125 なので、相対的なばらつきは A の方が大きい。
- 変動係数は A が 6.25、B が 8.0 なので、相対的なばらつきは B の方が大きい。
- 単位が同じなので、標準偏差の差 7 g だけで比べればよい。
- 平均が異なるので、ばらつきは比較できない。
