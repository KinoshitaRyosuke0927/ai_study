<!-- @question
type: choice
answer: 母集団の分布によらず、n が大きいとき標本平均は近似的に正規分布に従う。
title: 大数の法則と中心極限定理
hint: 中心極限定理は「標本平均の分布の形」について、大数の法則は「標本平均がどこに近づくか」についての定理です。
explanation: 中心極限定理は、母集団の分布によらず、n が大きければ標本平均が近似的に正規分布に従うことを述べています。大数の法則は、標本平均が母平均に近づくことを述べたもので、標本の分布が正規分布に近づくわけではありません。標本平均の分散は σ²/n で、n とともに小さくなります。
-->
大きさ $n$ の無作為標本の標本平均について述べた次の記述のうち、最も適切なものを選べ。

- n を大きくすると、標本(個々のデータ)の分布が正規分布に近づく。
- 母集団の分布によらず、n が大きいとき標本平均は近似的に正規分布に従う。
- 大数の法則によると、n を大きくすると標本平均の分散は σ² に近づく。
- 母集団が正規分布でないとき、標本平均の期待値は母平均と一致しない。
- n を大きくすると、標本平均の分散は大きくなる。

<!-- @question
type: choice
answer: I と II のみ
title: 推定量の性質
hint: 不偏性は期待値が母数に一致すること、有効性は不偏推定量の中で分散が小さいことです。
explanation: I. E[(X₁ + X₂)/2] = μ なので不偏推定量で正しい。II. 分散は X̄ が σ²/n、(X₁ + X₂)/2 が σ²/2 で、n > 2 なら X̄ の方が小さく、より有効なので正しい。III. 偏差の2乗和を n で割った標本分散は、期待値が (n − 1)σ²/n となり、母分散の不偏推定量ではないので誤り。
-->
母平均 $\mu$、母分散 $\sigma^2$ の母集団から、大きさ $n$($n > 2$)の無作為標本 $X_1, \dots, X_n$ を取り出す。次の記述 I〜III のうち、正しいものの組み合わせを選べ。

I. $(X_1 + X_2)/2$ は $\mu$ の不偏推定量である。

II. 標本平均 $\bar{X}$ は、$(X_1 + X_2)/2$ よりも分散が小さく、$\mu$ の推定量としてより有効である。

III. 偏差の2乗和を $n$ で割った標本分散は、$\sigma^2$ の不偏推定量である。

- I のみ
- II のみ
- I と II のみ
- I と III のみ
- I と II と III
