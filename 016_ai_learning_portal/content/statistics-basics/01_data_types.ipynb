## 概念

データは大きく **量的データ**(数値で大きさを表す)と **質的データ**(カテゴリを表す)に分けられます。
さらに、値の性質によって次の4つの **尺度水準** に分類されます。

| 尺度 | 例 | できる計算 |
|---|---|---|
| 名義尺度 | 血液型、クラス | 度数を数える |
| 順序尺度 | 満足度(1〜5)、順位 | 大小の比較 |
| 間隔尺度 | 気温(℃)、西暦 | 差をとる |
| 比例尺度 | 身長、点数、時間 | 比をとる |

尺度によって使える集計方法が変わるため、分析の最初にデータの種類を確認することが大切です。

## 演習

質的データは度数を数えて集計します。`class` 列について、クラスごとの人数をクラス名の順に表示してください。空欄 `____` を埋めて実行し、提出してください。

```python
# @starter
import pandas as pd

df = pd.read_csv("scores.csv")
counts = df["class"].____().sort_index()
print(counts)
```

```python
# @solution
# @title: クラスごとの人数
# @hint: Series の値ごとの出現回数は value_counts() で数えられます。
# @explanation: value_counts() は質的データの集計でよく使います。
import pandas as pd

df = pd.read_csv("scores.csv")
counts = df["class"].value_counts().sort_index()
print(counts)
```
